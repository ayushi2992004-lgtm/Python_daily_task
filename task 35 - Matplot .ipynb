```python
# 26. Create a monthly sales chart and identity the maximum-sales month. Highlight that point using annotate().
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20000, 25000, 22000, 30000, 28000, 35000]
plt.plot(months, sales, marker="o")
max_sales = max(sales)
max_month = months[sales.index(max_sales)]
plt.annotate(
    "Maximum Sales",
    xy=(max_month, max_sales),
    xytext=(max_month, max_sales + 3000),
    arrowprops=dict(arrowstyle="->")
)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
# 27. Create a scatter plot Age vs Income. Highlight customers whose income is greater than ₹1,00,000 using a different marker. 
age = [22, 25, 28, 30, 32, 35, 38, 40, 42, 45]
income = [50000, 70000, 90000, 110000, 80000, 120000, 95000, 150000, 130000, 75000]
normal_age = []
normal_income = []
high_age = []
high_income = []
for i in range(len(income)):
    if income[i] > 100000:
        high_age.append(age[i])
        high_income.append(income[i])
    else:
        normal_age.append(age[i])
        normal_income.append(income[i])
plt.scatter(normal_age, normal_income, marker="o", label="Income <= 100000")
plt.scatter(high_age, high_income, marker="^", label="Income > 100000")
plt.title("Age vs Income")
plt.xlabel("Age")
plt.ylabel("Income")
plt.legend()
plt.grid()
plt.show()
```

```python
# 28. Craete  a stacked bar chart showing monthly sales from:
# o Online
# o Offline
# o Corporate
months = ["Jan", "Feb", "Mar", "Apr", "May"]
online = [20, 25, 30, 35, 40]
offline = [15, 20, 25, 22, 30]
corporate = [10, 15, 20, 18, 25]
plt.bar(months, online, label="Online")
plt.bar(months, offline, bottom=online, label="Offline")
bottom = np.array(online) + np.array(offline)
plt.bar(months, corporate, bottom=bottom, label="Corporate")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# 29. Create a graph containing actual sales and target sales for 12 months. Highlight the months where actual sales are below the target.
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
actual = [80, 90, 75, 110, 95, 120, 100, 85, 130, 125, 115, 140]
target = [90, 90, 90, 100, 100, 110, 110, 100, 120, 130, 120, 130]
plt.plot(months, actual, marker="o", label="Actual Sales")
plt.plot(months, target, linestyle="--", label="Target Sales")
for i in range(len(months)):
    if actual[i] < target[i]:
        plt.scatter(months[i], actual[i], marker="x", s=100)
plt.title("Actual Sales vs Target Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid()
plt.show()
```
