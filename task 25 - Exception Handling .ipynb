```python
# 21. Create a custom exception InsufficientBalanceError for a banking program.
class InsufficientBalanceError(Exception):
    pass
balance = 5000

try:
    amount = int(input("Enter the Withraw Amount : "))
    if amount > balance:
        raise InsufficientBalanceError("Error : Insufficient Balance")

    balance = balance - amount
    print("Remaining Balance = ",balance)

except InsufficientBalanceError as e:
    print(e)
```

```python
# 26. Create a simple calculator that handles ZeroDivisionError, ValueError and invalid operator.
try:
    a = int(input("Enter First Number : "))
    b = int(input("Enter Second Number : "))
    op = input("Enter an Operator (+, -, *, /, %) : ")

    if op == "+":
        print("Addition = ", a+b)

    elif op == "-":
        print("Substraction = ", a-b)

    elif op == "*":
        print("Product = ", a*b)
        
    elif op == "/":
        print("Division = ", a/b)

    elif op == "%":
        print("Module = ", a%b)

    else:
        raise Exception("Error : Invalid Operator")

except ValueError:
    print("Error : Enter Valid Integer")

except ZeroDivisionError:
    print("Error : Division by Zero is not Allowed")

except Exception as e:
    print(e)
```

```python
# 28. Create a banking system with deposit and withdraw. Raise and handle custom exception for insufficient balance.
class InsufficientBalance(Exception):
    pass
balance = 10000

try:
    amount = int(input("Enter the Withdraw Amount : "))
    if amount > balance:
        raise InsufficientBalance("Error : Insufficient Balance, Please Try Again")

    balance = balance - amount
    print("Withdraw Successfully")
    print("Remaining Balance = ",balance)

except InsufficientBalanceError as e:
    print(e)
```

```python
# 30. Write a complete program that accepts user input for name, age and marks. Handle all possible exceptions and use finally to print “Program Ended”.
try:
    name = input("Enter Name : ")
    Age = int(input("Enter Age : "))
    if age<0:
        raise ValueError("Error : Invalid Age")

    marks = float(input("Enter Marks Obtained : "))
    if marks < 0 or marks > 100:
        raise Exception("Error : Enter Marks must be in between 0 and 100")

except ValueError as e:
    print(e)

except Exception as e:
    print(e)

else:
    print()
    print("\n------------- Student Details ------------------")
    print("Name: ", name)
    print("Age : ", age)
    print("Marks : ", marks)

finally:
    print("Program is Ended")
```
