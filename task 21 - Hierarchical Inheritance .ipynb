```python
# 1. Create a parent class Person with attributes name and age. Create two child classes Student and Teacher that inherit from Person.
# Parent Class
class Person:

    # Constructor
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    # Getters
    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    # Setters
    def set_name(self, name):
        self.__name = name
        return "Name Updated Successfully"

    def set_age(self, age):
        if age > 0:
            self.__age = age
            return "Age Updated Successfully"
        else:
            return "Invalid Age"

    # Display Method
    def show_person(self):
        print(f"Name : {self.__name}")
        print(f"Age  : {self.__age}")

# Child Class
class Student(Person):

    # Constructor
    def __init__(self, name, age, course):
        super().__init__(name, age)
        self.__course = course

    # Getter
    def get_course(self):
        return self.__course

    # Setter
    def set_course(self, course):
        self.__course = course
        return "Course Updated Successfully"

    # Display Method
    def show_student(self):
        self.show_person()
        print(f"Course : {self.__course}")

# Child Class
class Teacher(Person):

    # Constructor
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.__subject = subject

    # Getter
    def get_subject(self):
        return self.__subject
        
    # Setter
    def set_subject(self, subject):
        self.__subject = subject
        return "Subject Updated Successfully"

    # Display Method
    def show_teacher(self):
        self.show_person()
        print(f"Subject : {self.__subject}")

# Creating Objects
std = Student("Riya", 20, "Python")
teach = Teacher("Rahul", 35, "Data Science")

# Calling Methods
print("******** Student Details ********")
std.show_student()
print()
print("******** Teacher Details ********")
teach.show_teacher()
print()
print(std.set_course("AI & ML"))
print(teach.set_subject("Machine Learning"))
print()
print("Updated Student Course :", std.get_course())
print("Updated Teacher Subject :", teach.get_subject())
```

```python
# 6. Define class BankAccount with method deposit().
# Parent Class
class BankAccount:

    # Constructor
    def __init__(self, acc_no, balance):
        self.__acc_no = acc_no
        self.__balance = balance

    # Getters
    def get_acc_no(self):
        return self.__acc_no

    def get_balance(self):
        return self.__balance

    # Setters
    def set_acc_no(self, acc_no):
        self.__acc_no = acc_no
        return "Account Number Updated Successfully"

    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
            return "Balance Updated Successfully"
        else:
            return "Invalid Balance"

    # Deposit Method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return f"₹{amount} Deposited Successfully"
        else:
            return "Invalid Deposit Amount"

    # Display Method
    def show_account(self):
        print(f"Account Number : {self.__acc_no}")
        print(f"Balance        : ₹{self.__balance}")

# Child Class 1
class SavingsAccount(BankAccount):

    # Constructor
    def __init__(self, acc_no, balance, interest_rate):
        super().__init__(acc_no, balance)
        self.__interest_rate = interest_rate

    # Getter
    def get_interest_rate(self):
        return self.__interest_rate

    # Setter
    def set_interest_rate(self, interest_rate):
        self.__interest_rate = interest_rate
        return "Interest Rate Updated Successfully"

    # Display Method
    def show_savings(self):
        self.show_account()
        print(f"Interest Rate : {self.__interest_rate}%")

# Child Class 2
class CurrentAccount(BankAccount):

    # Constructor
    def __init__(self, acc_no, balance, overdraft_limit):
        super().__init__(acc_no, balance)
        self.__overdraft_limit = overdraft_limit

    # Getter
    def get_overdraft_limit(self):
        return self.__overdraft_limit

    # Setter
    def set_overdraft_limit(self, overdraft_limit):
        self.__overdraft_limit = overdraft_limit
        return "Overdraft Limit Updated Successfully"

    # Display Method
    def show_current(self):
        self.show_account()
        print(f"Overdraft Limit : ₹{self.__overdraft_limit}")


# Creating Objects
sa = SavingsAccount("SB1001", 50000, 6.5)
ca = CurrentAccount("CA2001", 75000, 100000)

# Calling Methods
print("********** Savings Account **********")
sa.show_savings()

print()
print("********** Current Account **********")
ca.show_current()
print()
print(sa.deposit(10000))
print(ca.deposit(25000))
print()
print(sa.set_interest_rate(7.0))
print(ca.set_overdraft_limit(150000))
print()
print("Updated Interest Rate :", sa.get_interest_rate(), "%")
print("Updated Overdraft Limit : ₹", ca.get_overdraft_limit())
```

```python
# 16. Polymorphism using overriding.
# Parent Class
class Shape:

    # Constructor
    def __init__(self, color):
        self.__color = color

    # Getter
    def get_color(self):
        return self.__color

    # Setter
    def set_color(self, color):
        self.__color = color
        return "Color Updated Successfully"

    # Display Method
    def area(self):
        print("Area of Shape")

    # Display Method
    def show_shape(self):
        print(f"Color : {self.__color}")


# Child Class 1
class Circle(Shape):

    # Constructor
    def __init__(self, color, radius):
        super().__init__(color)
        self.__radius = radius

    # Getter
    def get_radius(self):
        return self.__radius

    # Setter
    def set_radius(self, radius):
        if radius > 0:
            self.__radius = radius
            return "Radius Updated Successfully"
        else:
            return "Invalid Radius"

    # Method Overriding
    def area(self):
        area = 3.14 * self.__radius ** 2
        print(f"Area of Circle : {area}")

    # Display Method
    def show_circle(self):
        self.show_shape()
        print(f"Radius : {self.__radius}")
        self.area()


# Child Class 2
class Square(Shape):

    # Constructor
    def __init__(self, color, side):
        super().__init__(color)
        self.__side = side

    # Getter
    def get_side(self):
        return self.__side

    # Setter
    def set_side(self, side):
        if side > 0:
            self.__side = side
            return "Side Updated Successfully"
        else:
            return "Invalid Side"

    # Method Overriding
    def area(self):
        area = self.__side ** 2
        print(f"Area of Square : {area}")

    # Display Method
    def show_square(self):
        self.show_shape()
        print(f"Side : {self.__side}")
        self.area()


# Creating Objects
circle = Circle("Red", 7)
square = Square("Blue", 5)

# Calling Methods
print("********** Circle Details **********")
circle.show_circle()

print()

print("********** Square Details **********")
square.show_square()

print()

print(circle.set_radius(10))
print(square.set_side(8))

print()

print("Updated Radius :", circle.get_radius())
print("Updated Side :", square.get_side())

print()

print("********** Updated Areas **********")
circle.area()
square.area()
```

```python
# 18. Different child behavior.
# Parent Class
class BankAccount:

    # Constructor
    def __init__(self, acc_no, balance):
        self.__acc_no = acc_no
        self.__balance = balance

    # Getters
    def get_acc_no(self):
        return self.__acc_no

    def get_balance(self):
        return self.__balance

    # Setters
    def set_acc_no(self, acc_no):
        self.__acc_no = acc_no
        return "Account Number Updated Successfully"

    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
            return "Balance Updated Successfully"
        else:
            return "Invalid Balance"

    # Display Method
    def show_account(self):
        print(f"Account Number : {self.__acc_no}")
        print(f"Balance        : ₹{self.__balance}")

    # Method
    def calculate_interest(self):
        print("Interest Calculation")


# Child Class 1
class SavingsAccount(BankAccount):

    # Constructor
    def __init__(self, acc_no, balance, interest_rate):
        super().__init__(acc_no, balance)
        self.__interest_rate = interest_rate

    # Getter
    def get_interest_rate(self):
        return self.__interest_rate

    # Setter
    def set_interest_rate(self, interest_rate):
        self.__interest_rate = interest_rate
        return "Interest Rate Updated Successfully"

    # Method Overriding
    def calculate_interest(self):
        interest = (self.get_balance() * self.__interest_rate) / 100
        print(f"Savings Interest : ₹{interest}")

    # Display Method
    def show_savings(self):
        self.show_account()
        print(f"Interest Rate : {self.__interest_rate}%")
        self.calculate_interest()


# Child Class 2
class CurrentAccount(BankAccount):

    # Constructor
    def __init__(self, acc_no, balance, interest_rate):
        super().__init__(acc_no, balance)
        self.__interest_rate = interest_rate

    # Getter
    def get_interest_rate(self):
        return self.__interest_rate

    # Setter
    def set_interest_rate(self, interest_rate):
        self.__interest_rate = interest_rate
        return "Interest Rate Updated Successfully"

    # Method Overriding
    def calculate_interest(self):
        interest = (self.get_balance() * self.__interest_rate) / 100
        print(f"Current Interest : ₹{interest}")

    # Display Method
    def show_current(self):
        self.show_account()
        print(f"Interest Rate : {self.__interest_rate}%")
        self.calculate_interest()


# Creating Objects
sa = SavingsAccount("SB101", 50000, 6)
ca = CurrentAccount("CA201", 80000, 3)

# Calling Methods
print("********** Savings Account **********")
sa.show_savings()

print()

print("********** Current Account **********")
ca.show_current()

print()

print(sa.set_interest_rate(7))
print(ca.set_interest_rate(4))

print()

print("Updated Savings Interest Rate :", sa.get_interest_rate(), "%")
print("Updated Current Interest Rate :", ca.get_interest_rate(), "%")
```

```python
# 28. One parent with multiple children.
# Parent Class
class Shape:

    # Constructor
    def __init__(self, color):
        self.__color = color

    # Getter
    def get_color(self):
        return self.__color

    # Setter
    def set_color(self, color):
        self.__color = color
        return "Color Updated Successfully"

    # Display Method
    def show_shape(self):
        print(f"Color : {self.__color}")


# Child Class 1
class TwoDShape(Shape):

    # Constructor
    def __init__(self, color, length, breadth):
        super().__init__(color)
        self.__length = length
        self.__breadth = breadth

    # Getters
    def get_length(self):
        return self.__length

    def get_breadth(self):
        return self.__breadth

    # Setters
    def set_length(self, length):
        self.__length = length
        return "Length Updated Successfully"

    def set_breadth(self, breadth):
        self.__breadth = breadth
        return "Breadth Updated Successfully"

    # Child Specific Method
    def area(self):
        area = self.__length * self.__breadth
        print(f"Area of 2D Shape : {area}")

    # Display Method
    def show_2dshape(self):
        self.show_shape()
        print(f"Length  : {self.__length}")
        print(f"Breadth : {self.__breadth}")
        self.area()


# Child Class 2
class ThreeDShape(Shape):

    # Constructor
    def __init__(self, color, length, breadth, height):
        super().__init__(color)
        self.__length = length
        self.__breadth = breadth
        self.__height = height

    # Getters
    def get_height(self):
        return self.__height

    # Setters
    def set_height(self, height):
        self.__height = height
        return "Height Updated Successfully"

    # Child Specific Method
    def volume(self):
        volume = self.__length * self.__breadth * self.__height
        print(f"Volume of 3D Shape : {volume}")

    # Display Method
    def show_3dshape(self):
        self.show_shape()
        print(f"Length  : {self.__length}")
        print(f"Breadth : {self.__breadth}")
        print(f"Height  : {self.__height}")
        self.volume()


# Creating Objects
shape2d = TwoDShape("Red", 10, 5)
shape3d = ThreeDShape("Blue", 10, 5, 4)

# Calling Methods
print("********** 2D Shape Details **********")
shape2d.show_2dshape()

print()

print("********** 3D Shape Details **********")
shape3d.show_3dshape()

print()

print(shape2d.set_length(15))
print(shape2d.set_breadth(8))
print(shape3d.set_height(6))

print()

print("Updated Length :", shape2d.get_length())
print("Updated Breadth :", shape2d.get_breadth())
print("Updated Height :", shape3d.get_height())

print()

print("********** Updated Calculations **********")
shape2d.area()
shape3d.volume()
```

```python
# 30. Complete hierarchical inheritance program.
# Parent Class
class Hospital:

    # Constructor
    def __init__(self, hospital_name, location):
        self.__hospital_name = hospital_name
        self.__location = location

    # Getters
    def get_hospital_name(self):
        return self.__hospital_name

    def get_location(self):
        return self.__location

    # Setters
    def set_hospital_name(self, hospital_name):
        self.__hospital_name = hospital_name
        return "Hospital Name Updated Successfully"

    def set_location(self, location):
        self.__location = location
        return "Location Updated Successfully"

    # Display Method
    def show_hospital(self):
        print(f"Hospital Name : {self.__hospital_name}")
        print(f"Location      : {self.__location}")


# Child Class 1
class Doctor(Hospital):

    # Constructor
    def __init__(self, hospital_name, location, doctor_name, specialization):
        super().__init__(hospital_name, location)
        self.__doctor_name = doctor_name
        self.__specialization = specialization

    # Getters
    def get_doctor_name(self):
        return self.__doctor_name

    def get_specialization(self):
        return self.__specialization

    # Setters
    def set_doctor_name(self, doctor_name):
        self.__doctor_name = doctor_name
        return "Doctor Name Updated Successfully"

    def set_specialization(self, specialization):
        self.__specialization = specialization
        return "Specialization Updated Successfully"

    # Child Specific Method
    def treat_patient(self):
        print(f"Dr. {self.__doctor_name} is treating patients.")

    # Display Method
    def show_doctor(self):
        self.show_hospital()
        print(f"Doctor Name   : {self.__doctor_name}")
        print(f"Specialization: {self.__specialization}")
        self.treat_patient()


# Child Class 2
class Nurse(Hospital):

    # Constructor
    def __init__(self, hospital_name, location, nurse_name, shift):
        super().__init__(hospital_name, location)
        self.__nurse_name = nurse_name
        self.__shift = shift

    # Getters
    def get_nurse_name(self):
        return self.__nurse_name

    def get_shift(self):
        return self.__shift

    # Setters
    def set_nurse_name(self, nurse_name):
        self.__nurse_name = nurse_name
        return "Nurse Name Updated Successfully"

    def set_shift(self, shift):
        self.__shift = shift
        return "Shift Updated Successfully"

    # Child Specific Method
    def assist_doctor(self):
        print(f"{self.__nurse_name} is assisting the doctor.")

    # Display Method
    def show_nurse(self):
        self.show_hospital()
        print(f"Nurse Name    : {self.__nurse_name}")
        print(f"Shift         : {self.__shift}")
        self.assist_doctor()


# Child Class 3
class Patient(Hospital):

    # Constructor
    def __init__(self, hospital_name, location, patient_name, disease):
        super().__init__(hospital_name, location)
        self.__patient_name = patient_name
        self.__disease = disease

    # Getters
    def get_patient_name(self):
        return self.__patient_name

    def get_disease(self):
        return self.__disease

    # Setters
    def set_patient_name(self, patient_name):
        self.__patient_name = patient_name
        return "Patient Name Updated Successfully"

    def set_disease(self, disease):
        self.__disease = disease
        return "Disease Updated Successfully"

    # Child Specific Method
    def take_treatment(self):
        print(f"{self.__patient_name} is taking treatment.")

    # Display Method
    def show_patient(self):
        self.show_hospital()
        print(f"Patient Name  : {self.__patient_name}")
        print(f"Disease       : {self.__disease}")
        self.take_treatment()


# Creating Objects
doctor = Doctor("City Hospital", "Nagpur", "Rahul", "Cardiologist")
nurse = Nurse("City Hospital", "Nagpur", "Priya", "Morning")
patient = Patient("City Hospital", "Nagpur", "Ayushi", "Fever")

# Calling Methods
print("********** Doctor Details **********")
doctor.show_doctor()

print()

print("********** Nurse Details **********")
nurse.show_nurse()

print()

print("********** Patient Details **********")
patient.show_patient()

print()

print(doctor.set_specialization("Neurologist"))
print(nurse.set_shift("Night"))
print(patient.set_disease("Viral Fever"))

print()

print("Updated Specialization :", doctor.get_specialization())
print("Updated Shift :", nurse.get_shift())
print("Updated Disease :", patient.get_disease())
```
