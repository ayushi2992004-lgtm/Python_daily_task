```python
# 23. Create abstract class Shape with abstract method area(). Implement it in Circle and Rectangle using super().
from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):

    # Constructor
    def __init__(self, color):
        self.Color = color

    # Abstract Method
    @abstractmethod
    def area(self):
        pass

# Child Class
class Circle(Shape):

    # Constructor
    def __init__(self, color, radius):
        super().__init__(color)
        self.Radius = radius

    # Override Abstract Method
    def area(self):
        print("Color :", self.Color)
        print("Area of Circle :", 3.14 * self.Radius * self.Radius)

# Child Class
class Rectangle(Shape):

    # Constructor
    def __init__(self, color, length, width):
        super().__init__(color)
        self.Length = length
        self.Width = width

    # Override Abstract Method
    def area(self):
        print("Color :", self.Color)
        print("Area of Rectangle :", self.Length * self.Width)

# Create Objects
circle = Circle("Red", 7)
rectangle = Rectangle("Blue", 10, 5)

# Call Methods
circle.area()
print()
rectangle.area()
```

```python
# 6. Create multi-level inheritance: Person → Employee → Manager.
# Parent Class
class Person:

    # Constructor
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    # Getter Methods
    def get_name(self):
        return self.__name
        
    def get_age(self):
        return self.__age

    # Setter Methods
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Invalid Age")

    # Display Method
    def display(self):
        print("Name :", self.__name)
        print("Age :", self.__age)

# Child Class
class Employee(Person):

    # Constructor
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.__salary = salary

    # Getter
    def get_salary(self):
        return self.__salary

    # Setter
    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary
        else:
            print("Invalid Salary")

    # Override Method
    def display(self):
        super().display()
        print("Salary :", self.__salary)

# Grand Child Class
class Manager(Employee):

    # Constructor
    def __init__(self, name, age, salary, department):
        super().__init__(name, age, salary)
        self.__department = department

    # Getter
    def get_department(self):
        return self.__department

    # Setter
    def set_department(self, department):
        if department != "":
            self.__department = department
        else:
            print("Department Cannot be Empty")

    # Override Method
    def display(self):
        super().display()
        print("Department :", self.__department)

# Create Object
manager = Manager("Shruti", 21, 75000, "IT")

# Display Details
manager.display()
print()

# Update Values
manager.set_age(22)
manager.set_salary(85000)
manager.set_department("Data Science")
print()
print("Updated Details")
manager.display()
```

```python
# 8. Create hybrid inheritance structure and demonstrate Method Resolution Order (MRO). Also apply Duck Typing on the final child class.

# Parent Class
class Person:
    def work(self):
        print("Person is Working")

# Child Class 1
class Teacher(Person):
    def work(self):
        print("Teacher is Teaching")

# Child Class 2
class Programmer(Person):
    def work(self):
        print("Programmer is Coding")

# Hybrid Child Class
class Trainer(Teacher, Programmer):
    def training(self):
        print("Trainer is Taking Python Class")
    def start(self):
        print("Training Started")

# Another Class for Duck Typing
class Guest:
    def start(self):
        print("Guest Started Learning Python")

# Duck Typing Function
def begin(obj):
    obj.start()

# Create Object
t1 = Trainer()
print("------Hybrid Inheritance -----")
t1.work()          # MRO decides which work() method is called
t1.training()
print()

print("----- Duck Typing -----")
g1 = Guest()
begin(t1)
begin(g1)
print()

print("----- Method Resolution Order -----")
for i in Trainer.mro():
    print(i.__name__)
```

```python
# 9. Create a class ComplexNumber and overload +, -, *, and == operators completely.

# Create Class
class ComplexNumber:

    # Constructor
    def __init__(self, real, imaginary):
        self.Real = real
        self.Imaginary = imaginary

    # Overload + Operator
    def __add__(self, other):
        real = self.Real + other.Real
        imaginary = self.Imaginary + other.Imaginary
        return ComplexNumber(real, imaginary)

    # Overload - Operator
    def __sub__(self, other):
        real = self.Real - other.Real
        imaginary = self.Imaginary - other.Imaginary
        return ComplexNumber(real, imaginary)

    # Overload * Operator
    def __mul__(self, other):
        real = (self.Real * other.Real) - (self.Imaginary * other.Imaginary)
        imaginary = (self.Real * other.Imaginary) + (self.Imaginary * other.Real)
        return ComplexNumber(real, imaginary)

    # Overload == Operator
    def __eq__(self, other):
        return self.Real == other.Real and self.Imaginary == other.Imaginary

    # Display Method
    def display(self):
        print(self.Real, "+", self.Imaginary, "i")

# Create Objects
num1 = ComplexNumber(4, 3)
num2 = ComplexNumber(2, 5)
num3 = ComplexNumber(4, 3)

# Addition
print("Addition")
result = num1 + num2
result.display()
print()

# Subtraction
print("Subtraction")
result = num1 - num2
result.display()
print()

# Multiplication
print("Multiplication")
result = num1 * num2
result.display()
print()

# Equality
print("Equality Check")
if num1 == num3:
    print("Both Complex Numbers are Equal")
else:
    print("Both Complex Numbers are Not Equal")
```

```python
# 10. Create a class Student with private marks list.
# Create Class
class Student:

    # Constructor
    def __init__(self, name):
        self.Name = name
        self.__marks = []

    # Setter Method
    def set_marks(self, marks):
        valid_marks = []
        for i in marks:
            if i >= 0 and i <= 100:
                valid_marks.append(i)
            else:
                print(i, "is Invalid Mark")
        self.__marks = valid_marks

    # Getter Method
    def get_marks(self):
        return self.__marks

    # Average Method
    def average(self):
        total = 0
        for i in self.__marks:
            total = total + i
        avg = total / len(self.__marks)
        return avg

    # Operator Overloading
    def __gt__(self, other):
        return self.average() > other.average()

    # Display Method
    def display(self):
        print("Student Name :", self.Name)
        print("Marks :", self.__marks)
        print("Average :", self.average())

# Create Objects
student1 = Student("Riya")
student2 = Student("Rahul")

# Set Marks
student1.set_marks([85, 90, 88, 95, 91])
student2.set_marks([75, 80, 70, 82, 78])
print()

# Display Details
student1.display()
print()
student2.display()
print()

# Compare Students
if student1 > student2:
    print(student1.Name, "has Higher Average")
else:
    print(student2.Name, "has Higher Average")
```

```python
# 11. Take a number from user. Using while loop, reverse the number and check if it is a palindrome (without converting to string).
# Take Input
number = int(input("Enter a Number : "))

# Store Original Number
original = number

# Variable to Store Reverse Number
reverse = 0

# Reverse the Number
while number > 0:
    digit = number % 10
    reverse = (reverse * 10) + digit
    number = number // 10

# Display Results
print()
print("Original Number :", original)
print("Reversed Number :", reverse)
print()

# Check Palindrome
if original == reverse:
    print(original, "is a Palindrome Number")
else:
    print(original, "is Not a Palindrome Number")
```

```python
# 12. Create hierarchical inheritance: Parent Account → Children SavingsAccount and CurrentAccount.
#• Different interest calculation
#• Use abstract method for calculate_interest()
from abc import ABC, abstractmethod

# Parent Abstract Class
class Account(ABC):
    
    # Constructor
    def __init__(self, acc_no, name, balance):
        self.Account_No = acc_no
        self.Name = name
        self.Balance = balance

    # Display Method
    def display(self):
        print("Account Number :", self.Account_No)
        print("Account Holder :", self.Name)
        print("Balance :", self.Balance)

    # Abstract Method
    @abstractmethod
    def calculate_interest(self):
        pass

# Child Class
class SavingsAccount(Account):

    # Constructor
    def __init__(self, acc_no, name, balance):
        super().__init__(acc_no, name, balance)

    # Override Method
    def calculate_interest(self):
        interest = self.Balance * 5 / 100
        print("Savings Interest :", interest)

    # Display Method
    def display(self):
        print("----- Savings Account -----")
        super().display()

# Child Class
class CurrentAccount(Account):
    
    # Constructor
    def __init__(self, acc_no, name, balance):
        super().__init__(acc_no, name, balance)

    # Override Method
    def calculate_interest(self):
        interest = self.Balance * 2 / 100
        print("Current Interest :", interest)

    # Display Method
    def display(self):
        print("----- Current Account -----")
        super().display()

# Create Objects
s1 = SavingsAccount(101, "Manoj", 50000)
c1 = CurrentAccount(201, "Rahul", 80000)
print()

# Savings Account
s1.display()
s1.calculate_interest()
print()

# Current Account
c1.display()
c1.calculate_interest()
```

```python
# 13. Create a class Matrix and overload + and * operators for matrix addition and multiplication (with proper size validation).
# Create Class
class Matrix:

    # Constructor
    def __init__(self, matrix):
        self.matrix = matrix

    # Display Matrix
    def display(self):
        for row in self.matrix:
            print(row)

    # Overload + Operator
    def __add__(self, other):
        rows = len(self.matrix)
        cols = len(self.matrix[0])

        # Size Validation
        if rows != len(other.matrix) or cols != len(other.matrix[0]):
            print("Matrix Addition Not Possible")
            return None
        result = []
        for i in range(rows):
            temp = []
            for j in range(cols):
                temp.append(self.matrix[i][j] + other.matrix[i][j])
            result.append(temp)
        return Matrix(result)

    # Overload * Operator
    def __mul__(self, other):
        rows1 = len(self.matrix)
        cols1 = len(self.matrix[0])
        rows2 = len(other.matrix)
        cols2 = len(other.matrix[0])

        # Size Validation
        if cols1 != rows2:
            print("Matrix Multiplication Not Possible")
            return None
        result = []
        for i in range(rows1):
            temp = []
            for j in range(cols2):
                total = 0
                for k in range(cols1):
                    total = total + self.matrix[i][k] * other.matrix[k][j]
                temp.append(total)
            result.append(temp)
        return Matrix(result)

# Create Objects
m1 = Matrix([[1, 2], [3, 4]])
m2 = Matrix([[5, 6], [7, 8]])
print("Matrix 1")
m1.display()
print()
print("Matrix 2")
m2.display()
print()

# Addition
print("Matrix Addition")
add = m1 + m2
if add:
    add.display()
print()

# Multiplication
print("Matrix Multiplication")
mul = m1 * m2
if mul:
    mul.display()
```

```python
# 15. Create a class Time (hours, minutes, seconds) and overload + operator to add two time objects with proper carry-over.
# Create Class
class Time:

    # Constructor
    def __init__(self, hours, minutes, seconds):
        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds
        
    # Overload + Operator
    def __add__(self, other):

        # Add Seconds
        sec = self.seconds + other.seconds
        carry_min = sec // 60
        sec = sec % 60

        # Add Minutes
        minute = self.minutes + other.minutes + carry_min
        carry_hour = minute // 60
        minute = minute % 60

        # Add Hours
        hour = self.hours + other.hours + carry_hour
        
        # Return New Object
        return Time(hour, minute, sec)
        
    # Display Method
    def display(self):
        print(self.hours, "Hours",
              self.minutes, "Minutes",
              self.seconds, "Seconds")

# Create Object
time1 = Time(2, 45, 50)
time2 = Time(3, 30, 20)
print("Time 1")
time1.display()
print()
print("Time 2")
time2.display()
print()

# Add Time Objects
print("After Addition")
result = time1 + time2
result.display()
```

```python
# 16. Create abstract class Payment with abstract method process(). Implement CreditCard and UPI. Use Duck Typing to process any payment object.
from abc import ABC, abstractmethod

# Abstract Class
class Payment(ABC):

    # Constructor
    def __init__(self, amount):
        self.Amount = amount

    # Display Method
    def show_amount(self):
        print("Payment Amount :", self.Amount)

    # Abstract Method
    @abstractmethod
    def process(self):
        pass

# Child Class
class CreditCard(Payment):

    # Constructor
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.Card_Number = card_number
        
    # Override Method
    def process(self):
        print("Payment Done Using Credit Card")
        print("Card Number :", self.Card_Number)
        self.show_amount()

# Child Class
class UPI(Payment):

    # Constructor
    def __init__(self, amount, upi_id):
        super().__init__(amount)
        self.Upi_ID = upi_id

    # Override Method
    def process(self):
        print("Payment Done Using UPI")
        print("UPI ID :", self.Upi_ID)
        self.show_amount()

# Duck Typing Function
def make_payment(obj):
    obj.process()

# Create Objects
payment1 = CreditCard(2500, "1234-5678-9012")
payment2 = UPI(1500, "ayushi@upi")
print("------ Credit Card Payment ------")
make_payment(payment1)
print()
print("------ UPI Payment ------")
make_payment(payment2)
```

```python
# 17. Create a class Employee with private salary.
#• Setter should not allow more than 20% increase at once
#• Keep history of salary changes using class attribute
#• Add static method to calculate tax 

# Create Class
class Employee:
    
    # Class Attribute
    salary_history = []

    # Constructor
    def __init__(self, name, salary):
        self.Name = name
        self.__Salary = salary
        Employee.salary_history.append((self.Name, salary))

    # Getter Method
    def get_salary(self):
        return self.__Salary

    # Setter Method
    def set_salary(self, new_salary):
        increase = new_salary - self.__Salary
        max_increase = self.__Salary * 20 / 100
        if increase <= max_increase:
            self.__Salary = new_salary
            Employee.salary_history.append((self.Name, new_salary))
            print("Salary Updated Successfully")
        else:
            print("Salary Increase Cannot Exceed 20%")

    # Display Method
    def display(self):
        print("Employee Name :", self.Name)
        print("Salary :", self.__Salary)

    # Static Method
    @staticmethod
    def calculate_tax(salary):
        tax = salary * 10 / 100
        return tax

    # Class Method
    @classmethod
    def show_history(cls):
        print("\nSalary History")
        for name, salary in cls.salary_history:
            print(name, "→", salary)

# Create Objects
emp1 = Employee("Ayushi", 50000)
emp2 = Employee("Rahul", 40000)
print()

# Display Employee Details
emp1.display()
print()
emp2.display()
print()

# Update Salary
emp1.set_salary(58000)
emp2.set_salary(46000)
print()

# Display Updated Salary
emp1.display()
print()
emp2.display()
print()

# Calculate Tax
print("Tax of", emp1.Name, ":", Employee.calculate_tax(emp1.get_salary()))
print("Tax of", emp2.Name, ":", Employee.calculate_tax(emp2.get_salary()))

# Show Salary History
Employee.show_history()
```

```python
# 19. Create single, multi-level and hierarchical inheritance in one program and show the difference in method resolution.
# Parent class
class Person:

    # Constructor
    def __init__(self, name):
        self.Name = name

    # Method
    def show(self):
        print("Person Name :", self.Name)

    # Method
    def work(self):
        print("Person is Working")

# Single Inheritance 
class Student(Person):

    # Constructor
    def __init__(self, name, course):
        super().__init__(name)
        self.Course = course

    # Override Method
    def work(self):
        print("Student is Studying")

    # Display Method
    def display(self):
        super().show()
        print("Course :", self.Course)

# Multi-Level Inheritance
class Employee(Student):

    # Constructor
    def __init__(self, name, course, company):
        super().__init__(name, course)
        self.Company = company
        
    # Override Method
    def work(self):
        print("Employee is Working in", self.Company)

    # Display Method
    def display(self):
        super().display()
        print("Company :", self.Company)

# Hierarchical Inheritance 
class Teacher(Person):

    # Constructor
    def __init__(self, name, subject):
        super().__init__(name)
        self.Subject = subject

    # Override Method
    def work(self):
        print("Teacher is Teaching")

    # Display Method
    def display(self):
        super().show()
        print("Subject :", self.Subject)

class Doctor(Person):

    # Constructor
    def __init__(self, name, hospital):
        super().__init__(name)
        self.Hospital = hospital

    # Override Method
    def work(self):
        print("Doctor is Treating Patients")
        
    # Display Method
    def display(self):
        super().show()
        print("Hospital :", self.Hospital)

# Objects 
print("========== Single Inheritance ==========")
s1 = Student("Ayushi", "Python")
s1.display()
s1.work()
print()

print("========== Multi-Level Inheritance ==========")
e1 = Employee("Rahul", "Data Science", "TCS")
e1.display()
e1.work()
print()

print("========== Hierarchical Inheritance ==========")
t1 = Teacher("Priya", "Mathematics")
t1.display()
t1.work()
print()

d1 = Doctor("Amit", "City Hospital")
d1.display()
d1.work()
print()

# Method Resolution Order
print("========== Method Resolution Order ==========")
print("Student MRO")
for i in Student.mro():
    print(i.__name__)
print()
print("Employee MRO")
for i in Employee.mro():
    print(i.__name__)
print()
print("Teacher MRO")
for i in Teacher.mro():
    print(i.__name__)
print()
print("Doctor MRO")
for i in Doctor.mro():
    print(i.__name__)
```

```python
# 20. Create a complete system:
#• Abstract class Vehicle
#• Multiple inheritance with Electric and Fuel
#• Operator overloading for comparing vehicles based on speed
#• Duck Typing for start() method
#• Private attributes + getter/setter with validation

from abc import ABC, abstractmethod

# Abstract Class
class Vehicle(ABC):
    
    # Constructor
    def __init__(self, brand, speed):
        self.Brand = brand
        self.__Speed = speed
        
    # Getter
    def get_speed(self):
        return self.__Speed

    # Setter
    def set_speed(self, speed):
        if speed > 0:
            self.__Speed = speed
        else:
            print("Speed must be greater than 0")

    # Display Method
    def display(self):
        print("Brand :", self.Brand)
        print("Speed :", self.__Speed, "km/hr")

    # Abstract Method
    @abstractmethod
    def vehicle_type(self):
        pass
        
    # Operator Overloading
    def __gt__(self, other):
        return self.__Speed > other.__Speed

# Electric Class 
class Electric(Vehicle):
    def __init__(self, brand, speed, battery):
        Vehicle.__init__(self, brand, speed)
        self.Battery = battery
        
    def vehicle_type(self):
        print("Electric Vehicle")
        
    def charge(self):
        print("Battery Capacity :", self.Battery, "kWh")

# Fuel Class 
class Fuel(Vehicle):
    def __init__(self, brand, speed, fuel):
        Vehicle.__init__(self, brand, speed)
        self.Fuel = fuel
        
    def vehicle_type(self):
        print("Fuel Vehicle")
        
    def refill(self):
        print("Fuel Tank :", self.Fuel, "Litres")

# Multiple Inheritance
class HybridCar(Electric, Fuel):

    def __init__(self, brand, speed, battery, fuel):
        Vehicle.__init__(self, brand, speed)
        self.Battery = battery
        self.Fuel = fuel

    def vehicle_type(self):
        print("Hybrid Vehicle")

    def charge(self):
        print("Battery Capacity :", self.Battery, "kWh")

    def refill(self):
        print("Fuel Tank :", self.Fuel, "Litres")

    def start(self):
        print(self.Brand, "Started Successfully")

# Another Class for Duck Typing 
class Bike:
    def start(self):
        print("Bike Started Successfully")

# Duck Typing Function 
def start_vehicle(vehicle):
    vehicle.start()

# Objects 
car1 = HybridCar("Toyota", 180, 40, 35)
car2 = HybridCar("Tesla", 220, 75, 0)
print("----------- Vehicle 1 -----------")
car1.display()
car1.vehicle_type()
car1.charge()
car1.refill()
print()

print("----------- Vehicle 2 -----------")
car2.display()
car2.vehicle_type()
car2.charge()
car2.refill()
print()

# Operator Overloading
if car1 > car2:
    print(car1.Brand, "is Faster")
else:
    print(car2.Brand, "is Faster")
print()

# Getter & Setter 
print("Current Speed :", car1.get_speed())
car1.set_speed(200)
print("Updated Speed :", car1.get_speed())
print()

# Duck Typing 
print("Duck Typing")
bike = Bike()
start_vehicle(car1)
start_vehicle(bike)
```
