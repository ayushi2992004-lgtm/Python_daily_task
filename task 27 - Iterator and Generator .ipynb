```python
# 6. Create a custom iterator that prints numbers from 1 to 10.
class Numbers:
    def __iter__(self):
        self.num = 1
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

object = Numbers()
iterator = iter(object)

for i in iterator:
    print(i)
```

```python
# 8. Create a custom iterator that prints numbers in reverse order from 10 to 1. 
class ReverseNumbers:
    def __iter__(self):
        self.num = 10
        return self

    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value
        else:
            raise StopIteration

object = ReverseNumbers()
iterator = iter(object)

for i in iterator:
    print(i)
```

```python
# 9. Create a custom iterator that prints the characters of the word "GENERATOR" one by one.
class WordIterator:
    def __iter__(self):
        self.word = "GENERATOR"
        self.index = 0
        return self

    def __next__(self):
        if self.index < len(self.word):
            character = self.word[self.index]
            self.index += 1
            return character
        else:
            raise StopIteration

object = WordIterator()
iterator = iter(object)

for character in iterator:
    print(character)
```

```python
# 23. Create a custom iterator that prints all alphabet letters from A to Z.
class Alphabet:

    def __iter__(self):
        self.ch = 'A'
        return self

    def __next__(self):
        if self.ch <= 'Z':
            value = self.ch
            self.ch = chr(ord(self.ch) + 1)
            return value
        else:
            raise StopIteration

obj = Alphabet()
itr = iter(obj)

for i in itr:
    print(i)
```

```python
# 24. Create a custom iterator that prints all vowels from the string below. text = "Artificial Intelligence"
class Vowels:
    def __iter__(self):
        self.text = "Artificial Intelligence"
        self.index = 0
        return self

    def __next__(self):
        while self.index < len(self.text):

            ch = self.text[self.index]
            self.index += 1

            if ch.lower() in "aeiou":
                return ch

        raise StopIteration

obj = Vowels()
itr = iter(obj)

for i in itr:
    print(i)
```

```python
# 25. Create a custom iterator that prints the Fibonacci sequence up to 10 terms.
class Fibonacci:
    def __iter__(self):
        self.a = 0
        self.b = 1
        self.count = 1
        return self

    def __next__(self):
        if self.count <= 10:
            value = self.a
            c = self.a + self.b
            self.a = self.b
            self.b = c
            self.count += 1
            return value
        else:
            raise StopIteration

obj = Fibonacci()
itr = iter(obj)

for i in itr:
    print(i)
```

```python
# 26. Create a generator that yields the Fibonacci sequence up to 15 terms.
def fibonacci_numbers():
    a = 0
    b = 1
    for i in range(15):
        yield a
        c = a + b
        a = b
        b = c

gen = fibonacci_numbers()
for i in gen:
    print(i)
```

```python
# 27. Create a generator that yields all prime numbers between 1 and 100.
def prime_numbers():
    for num in range(1,100):
        prime = True
        for i in range(2,num):
            if num%i == 0:
                prime = False
                break
        if prime:
            yield num

gen = prime_numbers()
for i in gen:
    print(i)
```

```python
# 28. Create a generator that yields all factors of a given number. Example Input: num = 24
num = 24
def factors():
    for i in range(1, num+1):
        if num % i == 0:
            yield i

gen = factors()
for i in gen:
    print(i)
```

```python
# 30. Create an infinite generator that yields multiples of 5. Print only the first 15 values using next().
def multiple():
    num = 5
    while True:
        num = num + 5
        yield num

gen = multiple()
for i in range(15):
    print(next(gen))
```
