```python
# 6. Create a parent class named BankAccount provide methods deposit() and withdraw(). Create a subclass SavingAccount and add calculate_interest().
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Amount Deposited successfully")
        print(f"Balance = { self.balance}")

    def withdraw(self, amount):
        self.balance -= amount
        print("Amount Withdraw successfully")
        print(f"Balance: {self.balance}")

class SavingsAccount(BankAccount):
    def calculate_interest(self):
        print("Interest =", self.balance * 0.05)

s = SavingsAccount(10000)
s.deposit(2000)
s.withdraw(1000)
s.calculate_interest()
```

```python
# 29. Create multi-level inheritance Shape → TwoDShape → Triangle. Triangle should calculate its area using Heron's Formula.
class Shape:
    pass

class TwoDShape(Shape):
    pass

class Triangle(TwoDShape):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def area(self):
        s = (self.a + self.b + self.c) / 2
        area = (s * (s-self.a) * (s-self.b) * (s-self.c)) ** 0.5
        print("Area =", area)

t = Triangle(3, 4, 5)
t.area()
```
