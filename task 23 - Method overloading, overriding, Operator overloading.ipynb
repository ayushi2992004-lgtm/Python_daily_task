```python
# 1. Create a Calculator clss that can add 2 numbers or 3 numbers using the same method name. 
# Method Overloading using *args.

# Parent class
class Calculator:

    # Constructor
    def __init__(self, brand):
        self.Brand = brand
        print("Calculator Brand :", self.Brand)

    # Method Overloading
    def add(self, *numbers):
        result = 0

        for num in numbers:
            result += num
        return result

# Create Object
c1 = Calculator("Casio")

print("-" * 40)
print(c1.add(10, 20))
print(c1.add(10, 20, 30))
print(c1.add(10, 20, 30, 40))
```

```python
# 2. Create an Area class that Calculate the area of a square and rectangle using the same method.
# Parent class
class Area:

    # Constructor
    def __init__(self, shape):
        self.Shape = shape
        print("Shape :", self.Shape)

    # Method Overloading
    def calculate(self, length, width=None):

        # Square
        if width == None:
            result = length * length
            return result

        # Rectangle
        else:
            result = length * width
            return result

# Create Object
a1 = Area("Square & Rectangle")
print("-" * 40)

# Square
print("Area of Square :", a1.calculate(5))

# Rectangle
print("Area of Rectangle :", a1.calculate(5, 10))
```

```python
# 3. Create a Student class with a details() method that accepts defferent number of arguments.
# Parent class
class Student:

    # Constructor
    def __init__(self, college):
        self.College = college
        print("College :", self.College)

    # Method Overloading
    def details(self, name, age=None, city=None):

        # Name Only
        if age == None and city == None:
            print("Student Name :", name)

        # Name and Age
        elif city == None:
            print("Student Name :", name)
            print("Student Age :", age)

        # Name, Age and City
        else:
            print("Student Name :", name)
            print("Student Age :", age)
            print("Student City :", city)

# Create Object
s1 = Student("PBCOE Nagpur")
print("-" * 40)

# Name Only
s1.details("Ananya")
print("-" * 40)

# Name and Age
s1.details("Ananya", 21)
print("-" * 40)

# Name, Age and City
s1.details("Ananya", 21, "Nagpur")
```

```python
# 4. Create a MathOperation class that multiplies any number of values.
class MathOperations:

    # Constructor
    def __init__(self, operation):
        self.Operation = operation
        print("Operation :", self.Operation)

    # Method Overloading using *args
    def multiply(self, *numbers):
        result = 1

        for num in numbers:
            result = result * num
        return result

# Create Object
m1 = MathOperations("Multiplication")

# 2 Numbers
print("Multiply 2 Numbers :", m1.multiply(10, 20))

# 3 Numbers
print("Multiply 3 Numbers :", m1.multiply(10, 20, 30))

# 4 Numbers
print("Multiply 4 Numbers :", m1.multiply(10, 20, 30, 40))

# 5 Numbers
print("Multiply 5 Numbers :", m1.multiply(10, 20, 30, 40, 50))
```

```python
# 5. Create a Display class that prints defferent meassages based on the argument type.
# Parent class
class Display:

    # Constructor
    def __init__(self, title):
        self.Title = title
        print("Title :", self.Title)

    # Method Overloading
    def show(self, data):

        # Integer
        if isinstance(data, int):
            print("Integer Value :", data)

        # String
        elif isinstance(data, str):
            print("String Value :", data)

        # List
        elif isinstance(data, list):
            print("List Values :")
            for value in data:
                print(value)

        # Other Data Types
        else:
            print("Invalid Data Type")

# Create Object
d1 = Display("Display Data")

# Integer
d1.show(100)

# String
d1.show("Python")

# List
d1.show(["Python", "Java", "C++"])

# Float
d1.show(10.5)
```

```python
# 6. Create a Volume class that calculate the volume of a cube and cuboid.
# Parent class
class Volume:

    # Constructor
    def __init__(self, shape):
        self.Shape = shape
        print("Shape :", self.Shape)

    # Method Overloading
    def calculate(self, length, width=None, height=None):

        # Cube
        if width == None and height == None:
            result = length * length * length
            return result

        # Cuboid
        else:
            result = length * width * height
            return result

# Create Object
v1 = Volume("Cube")
v2 = Volume("Cuboid")

# Cube
print("Volume of Cube :", v1.calculate(5))

# Cuboid
print("Volume of Cuboid :", v1.calculate(5, 4, 3))
```

```python
# 7. Create a PrintData class that prints one or many values.
# Parent class
class PrintData:

    # Constructor
    def __init__(self, title):
        self.Title = title
        print("Title :", self.Title)

    # Method Overloading using *args
    def print_data(self, *args):

        print("Printing Data :")
        for value in args:
            print(value)

# Create Object
p1 = PrintData("Student Information")
print("-"*30)

# One Value
p1.print_data("Ayushi")

# Multiple Values
p1.print_data("Shreya",21,"Nagpur","Python","Data Science")
```

```python
# 8. Create an Animal class with sound(). Override it in Dog and Cat.
# Method Overriding
# Same Method - Different Implementation

# Parent class
class Animal:

    # Constructor
    def __init__(self, name):
        self.Name = name

    # Method 1
    def sound(self):
        print(f"{self.Name} makes Sound")

    # Method 2
    def move(self):
        print(f"{self.Name} is Moving")

# Override the methods using Child Class
# Child Class 1
class Cat(Animal):
    def sound(self):
        print(f"{self.Name} Sound : Meow Meow")

    def move(self):
        print(f"{self.Name} is Running")

# Child Class 2
class Dog(Animal):
    def sound(self):
        print(f"{self.Name} Sound : Woof Woof")

    def move(self):
        print(f"{self.Name} is Attacking")

# Create always Child Class Object
cat = Cat("Kitty")
dog = Dog("Bruno")

# Creating for loop for calling methods
animals = [cat, dog]
for animal in animals:
    animal.sound()
    animal.move()
```

```python
# 9. Create Vehicle, Car, and Bike classes.
# Parent Class
class Vehicle:
    # Constructor
    def __init__(self, name):
        self.Name = name

    # Method 1
    def start(self):
        print(f"{self.Name} Vehicle Started")

    # Method 2
    def fuel(self):
        print(f"{self.Name} uses Fuel")

# Override the methods using Child Class
# Child Class 1
class Car(Vehicle):
    def start(self):
        print(f"{self.Name} Car Started with Push Button")

    def fuel(self):
        print(f"{self.Name} uses Petrol")

# Child Class 2
class Bike(Vehicle):
    def start(self):
        print(f"{self.Name} Bike Started with Self Start")

    def fuel(self):
        print(f"{self.Name} uses Petrol")

# Create always Child Class Object
car = Car("Toyota Fortuner")
bike = Bike("Royal Enfield")

# Creating for loop for calling methods
vehicles = [car, bike]
for vehicle in vehicles:
    vehicle.start()
    vehicle.fuel()
```

```python
# 10. Create an Employee class and override calculate_salary() in manager.
# Parent class
class Employee:
    # Constructor
    def __init__(self, name, salary):
        self.Name = name
        self.Salary = salary

    # Method 1
    def calculate_salary(self):
        print(f"{self.Name} Salary : {self.Salary}")

    # Method 2
    def work(self):
        print(f"{self.Name} is Working")

# Override the methods using Child Class
# Child Class 1
class Manager(Employee):
    def calculate_salary(self):
        bonus = 20000
        print(f"{self.Name} Salary : {self.Salary + bonus}")

    def work(self):
        print(f"{self.Name} is Managing the Team")

# Child Class 2
class Developer(Employee):
    def calculate_salary(self):
        incentive = 10000
        print(f"{self.Name} Salary : {self.Salary + incentive}")

    def work(self):
        print(f"{self.Name} is Developing Software")

# Create always Child Class Object
manager = Manager("Preeti", 80000)
developer = Developer("Shyam", 60000)

# Creating for loop for calling methods
employees = [manager, developer]
for employee in employees:
    employee.calculate_salary()
    employee.work()
```

```python
# 11. Create Shape, Circle, and Rectangle classes.
# Parent class
class Shape:

    # Constructor
    def __init__(self, name):
        self.Name = name

    # Method 1
    def area(self):
        print(f"{self.Name} Area")

    # Method 2
    def draw(self):
        print(f"{self.Name} Shape Draw")

# Override the methods using Child Class
# Child Class 1
class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.Radius = radius

    def area(self):
        print(f"{self.Name} Area : {3.14 * self.Radius * self.Radius}")
        
    def draw(self):
        print(f"{self.Name} is Drawn")

# Child Class 2
class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__("Rectangle")
        self.Length = length
        self.Width = width

    def area(self):
        print(f"{self.Name} Area : {self.Length * self.Width}")

    def draw(self):
        print(f"{self.Name} is Drawn")

# Create always Child Class Object
circle = Circle(7)
rectangle = Rectangle(10, 5)

# Creating for loop for calling methods
shapes = [circle, rectangle]
for shape in shapes:
    shape.area()
    shape.draw()
```

```python
# 12. Create BankAccount, SavingAccount, and CurrentAccount.
# Parent class
class BankAccount:

    # Constructor
    def __init__(self, holder):
        self.Holder = holder

    # Method 1
    def interest(self):
        print(f"{self.Holder} Interest Rate : 3%")

    # Method 2
    def account_type(self):
        print(f"{self.Holder} has a Bank Account")

# Override the methods using Child Class
# Child Class 1
class SavingsAccount(BankAccount):
    def interest(self):
        print(f"{self.Holder} Interest Rate : 6%")

    def account_type(self):
        print(f"{self.Holder} has a Savings Account")

# Child Class 2
class CurrentAccount(BankAccount):
    def interest(self):
        print(f"{self.Holder} Interest Rate : 0%")

    def account_type(self):
        print(f"{self.Holder} has a Current Account")

# Create always Child Class Object
saving = SavingsAccount("Riya")
current = CurrentAccount("Preet")

# Creating for loop for calling methods
accounts = [saving, current]
for account in accounts:
    account.interest()
    account.account_type()
```

```python
# 13. Create Parent and Child classes using super().
# Parent class
class Employee:

    # Constructor
    def __init__(self, name, department):
        self.Name = name
        self.Department = department

    # Method
    def display(self):
        print(f"Employee Name : {self.Name}")
        print(f"Department : {self.Department}")

# Override the method using Child Class
class Manager(Employee):

    # Constructor
    def __init__(self, name, department, team):
        super().__init__(name, department)
        self.Team = team

    # Overriding Method
    def display(self):

        # Calling Parent Method
        super().display()

        # Child Class Information
        print(f"Managing Team : {self.Team}")

# Create Child Class Object
manager = Manager("Shrawai","Data Science","AI Development Team")

# Call Method
manager.display()
```

```python
# 14. Create Bird and Penguin Classes.
# Parent class
class Bird:

    # Constructor
    def __init__(self, name):
        self.Name = name

    # Method 1
    def fly(self):
        print(f"{self.Name} can Fly")

    # Method 2
    def food(self):
        print(f"{self.Name} eats Seeds")

# Override the methods using Child Class
# Child Class 1
class Eagle(Bird):
    def fly(self):
        print(f"{self.Name} flies at High Altitude")

    def food(self):
        print(f"{self.Name} eats Meat")

# Child Class 2
class Penguin(Bird):
    def fly(self):
        print(f"{self.Name} cannot Fly")
        
    def food(self):
        print(f"{self.Name} eats Fish")

# Create always Child Class Object
eagle = Eagle("Golden Eagle")
penguin = Penguin("Emperor Penguin")

# Creating for loop for calling methods
birds = [eagle, penguin]
for bird in birds:
    bird.fly()
    bird.food()
```

```python
# 15. Create a Point class and overload the + operator.
# Parent class
class Point:

    # Constructor
    def __init__(self, x, y):
        self.X = x
        self.Y = y

    # Dunder Method : __add__
    def __add__(self, other):
        x = self.X + other.X
        y = self.Y + other.Y
        return Point(x, y)

    # Dunder Method : __repr__
    def __repr__(self):
        return f"Point(X = {self.X}, Y = {self.Y})"

# Create Objects
p1 = Point(10, 20)
p2 = Point(30, 40)

# Operator Overloading
result = p1 + p2
print("Point 1 :", p1)
print("Point 2 :", p2)
print("Added Point :", result)
```

```python
# 16. Create a Book class overload the + operator.
# Parent Class
class Book:

    # Constructor
    def __init__(self, title, pages):
        self.Title = title
        self.Pages = pages

    # Dunder Method : __add__
    def __add__(self, other):
        total_pages = self.Pages + other.Pages
        return total_pages

    # Dunder Method : __repr__
    def __repr__(self):
        return f"Book(Title = '{self.Title}', Pages = {self.Pages})"

# Create Objects
b1 = Book("Python Programming", 350)
b2 = Book("Data Science", 450)

# Operator Overloading
print("Book 1 :", b1)
print("Book 2 :", b2)
print("Total Pages :", b1 + b2)
```

```python
# 17. Create a 'student class and overload the > operator.
# Parent class
class Student:

    # Constructor
    def __init__(self, name, marks):
        self.Name = name
        self.Marks = marks

    # Dunder Method : __gt__
    def __gt__(self, other):
        return self.Marks > other.Marks

    # Dunder Method : __repr__
    def __repr__(self):
        return f"Student(Name = '{self.Name}', Marks = {self.Marks})"

# Create Objects
s1 = Student("Sumit", 85)
s2 = Student("Suhani", 78)

# Operator Overloading
print("Student 1 :", s1)
print("Student 2 :", s2)
print()
if s1 > s2:
    print(f"{s1.Name} has scored more marks.")
else:
    print(f"{s2.Name} has scored more marks.")
```

```python
# 18. Create a Rectangle class and overload the == operator.
# Parent class
class Rectangle:

    # Constructor
    def __init__(self, length, width):
        self.Length = length
        self.Width = width

    # Dunder Method : __eq__
    def __eq__(self, other):
        area1 = self.Length * self.Width
        area2 = other.Length * other.Width
        return area1 == area2

    # Dunder Method : __repr__
    def __repr__(self):
        return f"Rectangle(Length = {self.Length}, Width = {self.Width})"

# Create Objects
r1 = Rectangle(10, 5)
r2 = Rectangle(25, 2)

# Operator Overloading
print("Rectangle 1 :", r1)
print("Rectangle 2 :", r2)
print()
if r1 == r2:
    print("Both Rectangles have the Same Area.")
else:
    print("Both Rectangles have Different Areas.")
```

```python
# 19. Create a Money class and overload the + and - Operators.
# Parent Class
class Money:

    # Constructor
    def __init__(self, amount):
        self.Amount = amount

    # Dunder Method : __add__
    def __add__(self, other):
        total = self.Amount + other.Amount
        return Money(total)

    # Dunder Method : __sub__
    def __sub__(self, other):
        balance = self.Amount - other.Amount
        return Money(balance)

    # Dunder Method : __repr__
    def __repr__(self):
        return f"Money(Amount = ₹{self.Amount})"

# Create Objects
m1 = Money(5000)
m2 = Money(2000)

# Operator Overloading
print("Wallet 1 :", m1)
print("Wallet 2 :", m2)
print()
print("Total Money :", m1 + m2)
print("Remaining Money :", m1 - m2)
```

```python
# 20. Create a Distance class and overload <, > and ==.
# Parent class
class Distance:

    # Constructor
    def __init__(self, kilometer):
        self.Kilometer = kilometer

    # Dunder Method : __lt__
    def __lt__(self, other):
        return self.Kilometer < other.Kilometer

    # Dunder Method : __gt__
    def __gt__(self, other):
        return self.Kilometer > other.Kilometer

    # Dunder Method : __eq__
    def __eq__(self, other):
        return self.Kilometer == other.Kilometer

    # Dunder Method : __repr__
    def __repr__(self):
        return f"Distance({self.Kilometer} km)"

# Create Objects
d1 = Distance(120)
d2 = Distance(150)

# Operator Overloading
print("Distance 1 :", d1)
print("Distance 2 :", d2)
print()
print("d1 < d2 :", d1 < d2)
print("d1 > d2 :", d1 > d2)
print("d1 == d2 :", d1 == d2)
```
