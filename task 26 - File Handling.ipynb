```python
# 22. Write a program that reads a log file and counts the number of “ERROR” messages.
try:
    file = open("log.txt", "r")
    count = 0
    for line in file:
        if "ERROR" in line:
            count += 1
    print("Number of ERROR messages =", count)
    file.close()

except FileNotFoundError:
    print("Error: File does not exist.")

except PermissionError:
    print("Error: Permission denied.")

except Exception as e:
    print("An unexpected error occurred:", e)
```

```python
# 25. Read a file containing Student marks, calculate average, and write the result in another file.
file = open("marks.txt","r")
marks = file.readlines()
total = 0
for mark in marks:
    total = total + int(mark.strip())
average = total/ len(marks)
result = open("average.txt","w")
result.write("Total Marks = "+ str(total)+"\n")
result.write("Average = "+str(average))
result.close()
print("Average Calculated Successfully")
```

```python
# 26. Create a program that keeps a record of daily expenses in a file and can show total expenses.
n = int(input("Enter number of expenses: "))
with open("expenses.txt", "w") as file:
    for i in range(n):
        expense = input("Enter expense: ")
        file.write(expense + "\n")

with open("expenses.txt", "r") as file:
    total = 0
    for amount in file:
        total = total + int(amount)

print("Total Expenses =", total)
```

```python
# 27. Write a program to remove all blank lines from a file.
with open("file1.txt", "r") as file:
    lines = file.readlines()

with open("file2.txt", "w") as file:
    for line in lines:
        if line.strip() != "":
            file.write(line)
print("Blank lines removed successfully.")
```

```python
# 28. Read a file and write only the unique lines into another file (remove duplicates).
with open("file1.txt", "r") as file:
    lines = file.readlines()

unique_lines = []

with open("file2.txt", "w") as file:
    for line in lines:
        if line not in unique_lines:
            unique_lines.append(line)
            file.write(line)

print("Duplicate lines removed successfully.")
```

```python
# 29.Create a menu-driven program:
# • Write to file
# • Read file
# • Append to file
# • Exit
while True:
    print("\n----- MENU -----")
    print("1. Write to File")
    print("2. Read File")
    print("3. Append to File")
    print("4. Exit")
    choice = int(input("Enter your choice: "))

    if choice == 1:
        text = input("Enter text to write: ")

        with open("data.txt", "w") as file:
            file.write(text)
        print("Data written successfully.")
        
    elif choice == 2:
        with open("data.txt", "r") as file:
            print("\nFile Content:")
            print(file.read())

    elif choice == 3:
        text = input("Enter text to append: ")

        with open("data.txt", "a") as file:
            file.write("\n" + text)
        print("Data appended successfully.")

    elif choice == 4:
        print("Program Exited.")
        break

    else:
        print("Invalid Choice!")
```
