```python
# 26.  Create class BankAccount with private __balance. Add deposit and withdraw methods. Raise custom exception if balance becomes negative.
class NegativeBalanceError(Exception):
    pass

class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    # Deposit
    def deposit(self,amount):
        self._amount = amount
        balance = self.__balance + amount
        print("Amount Deposited = ", self._amount)

    # Withdraw
    def withdraw(self,amount):
        self._amount = amount
        if amount > self.__balance:
            raise NegativeBalanceError("Error : Insufficient Balance, Please try Again.")
        else:
            balance = self.__balance - amount
            print("Amount Withdraw = ", self._amount)

    # getter
    def get_balance(self):
        return self.__balance

# object
B = BankAccount(5000)

try:
    B.deposit(2000)
    B.withdraw(1000)
    B.withdraw(5000)

except NegativeBalanceError as e:
    print(e)
    
print("Current Balance = ", B.get_balance())
```

```python
# 29. Create class Rectangle with private length and width. Add method to calculate area and perimeter.
class Rectangle:

    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    # Method 
    def area(self):
        return self.__length * self.__width

    # Method
    def perimeter(self):
        return 2 * (self.__length + self.__width)

length = int(input("Enter Length: "))
width = int(input("Enter Width: "))

r = Rectangle(length, width)

print("Area =", r.area())
print("Perimeter =", r.perimeter())
```

```python
# 33. Create abstract class Shape with abstract method area(). Implement it in Circle and Square using super().
from abc import ABC, abstractmethod

class Shape(ABC):

    def __init__(self):
        print("Shape Constructor")

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):

    def __init__(self, radius):
        super().__init__()
        self.radius = radius

    def area(self):
        print("Area of Circle =", 3.14 * self.radius * self.radius)

class Square(Shape):

    def __init__(self, side):
        super().__init__()
        self.side = side

    def area(self):
        print("Area of Square =", self.side * self.side)

c = Circle(5)
c.area()

s = Square(4)
s.area()
```

```python
# 41. Create a class Student with private marks list.
# • Setter should validate each mark (0-100)
# • Overload > operator to compare two students based on average
# • Raise custom exception if marks list is empty
class EmptyMarksError(Exception):
    pass

class Student:

    def __init__(self):
        self.__marks = []

    def set_marks(self, marks):

        if len(marks) == 0:
            raise EmptyMarksError("Marks list is empty")

        for i in marks:
            if i < 0 or i > 100:
                print("Invalid Marks")
                return

        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def average(self):
        return sum(self.__marks) / len(self.__marks)

    def __gt__(self, other):
        return self.average() > other.average()


try:

    s1 = Student()
    s2 = Student()

    s1.set_marks([80, 90, 85])
    s2.set_marks([70, 75, 80])

    print("Student 1 Marks:", s1.get_marks())
    print("Student 2 Marks:", s2.get_marks())

    print("Average of Student 1:", s1.average())
    print("Average of Student 2:", s2.average())

    if s1 > s2:
        print("Student 1 has higher average")
    else:
        print("Student 2 has higher average")

except EmptyMarksError as e:
    print(e)
```

```python
# 43. Using dictionary + exception handling, create a student management system that can:
# • Add student
# • Update marks
# • Delete student
# • Display all Raise custom exception if student not found.
class StudentNotFoundError(Exception):
    pass

students = {}

while True:
    print("\n1. Add Student")
    print("2. Update Marks")
    print("3. Delete Student")
    print("4. Display Students")
    print("5. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        name = input("Enter Student Name: ")
        marks = int(input("Enter Marks: "))
        students[name] = marks
        print("Student Added Successfully")

    elif choice == 2:
        try:
            name = input("Enter Student Name: ")
            if name not in students:
                raise StudentNotFoundError
            marks = int(input("Enter New Marks: "))
            students[name] = marks
            print("Marks Updated Successfully")

        except StudentNotFoundError:
            print("Student Not Found")

    elif choice == 3:
        try:
            name = input("Enter Student Name: ")
            if name not in students:
                raise StudentNotFoundError

            del students[name]
            print("Student Deleted Successfully")

        except StudentNotFoundError:
            print("Student Not Found")

    elif choice == 4:
        print("Student Records")
        for name, marks in students.items():
            print(name, ":", marks)

    elif choice == 5:
        print("Program Ended")
        break

    else:
        print("Invalid Choice")
```

```python
# 44. Create a class Employee with private salary.
# • Setter should not allow more than 15% increase at once
# • Keep history of all salary changes
# • Add static method to calculate tax
# • Overload > operator to compare employees by salary

class Employee:
    def __init__(self, salary):
        self.__salary = salary
        self.history = [salary]

    # Getter
    def get_salary(self):
        return self.__salary

    # Setter
    def set_salary(self, salary):
        increase = salary - self.__salary
        if increase > (self.__salary * 15 / 100):
            print("Salary increase cannot be more than 15%")
        else:
            self.__salary = salary
            self.history.append(salary)
            print("Salary Updated")

    # Salary
    def show_history(self):
        print("Salary History:", self.history)

    # Static Method
    @staticmethod
    def calculate_tax(salary):
        tax = salary * 10 / 100
        return tax

    # Operator Overloading
    def __gt__(self, other):
        return self.__salary > other.__salary


e1 = Employee(50000)
e2 = Employee(45000)

print("Employee 1 Salary:", e1.get_salary())
print("Employee 2 Salary:", e2.get_salary())

e1.set_salary(55000)
e1.show_history()
print("Tax =", Employee.calculate_tax(e1.get_salary()))

if e1 > e2:
    print("Employee 1 has higher salary")
else:
    print("Employee 2 has higher salary")
```

```python
# 45. Design a complete system combining:
# • Abstract class Vehicle
# • Multiple inheritance (Electric + Fuel)
# • Operator overloading (compare by speed)
# • Private attributes + getter/setter
# • Custom exceptions
# • File handling to save vehicle details
# • Duck Typing for start() method

from abc import ABC, abstractmethod

# Custom Exception
class InvalidSpeedError(Exception):
    pass

# Abstract Class
class Vehicle(ABC):

    @abstractmethod
    def display(self):
        pass

# Parent Class 1
class Electric:
    def charge(self):
        print("Charging Electric Vehicle")


# Parent Class 2
class Fuel:
    def fill(self):
        print("Filling Fuel")

# Child Class
class Car(Vehicle, Electric, Fuel):
    def __init__(self, name, speed):
        self.name = name
        self.__speed = 0
        self.set_speed(speed)

    # Getter
    def get_speed(self):
        return self.__speed

    # Setter
    def set_speed(self, speed):
        if speed < 0:
            raise InvalidSpeedError("Invalid Speed")
        self.__speed = speed

    # Abstract Method
    def display(self):
        print("Vehicle Name:", self.name)
        print("Speed:", self.__speed)

    # Operator Overloading
    def __gt__(self, other):
        return self.__speed > other.__speed


# Duck Typing Classes
class Bike:

    def start(self):
        print("Bike Started")

class Bus:

    def start(self):
        print("Bus Started")

def start_vehicle(vehicle):
    vehicle.start()

try:

    c1 = Car("Tesla", 180)
    c2 = Car("BMW", 160)

    c1.display()
    c1.charge()
    c1.fill()

    if c1 > c2:
        print("Tesla is Faster")
    else:
        print("BMW is Faster")

    file = open("vehicle.txt", "w")
    file.write(c1.name + " " + str(c1.get_speed()))
    file.close()

    b1 = Bike()
    b2 = Bus()

    start_vehicle(b1)
    start_vehicle(b2)

except InvalidSpeedError as e:
    print(e)
```
