```python
# 1. Create a hybride inheritance structure with classes A → B and C → D(Where D inherits from both B and C).
# Hybrid Inheritance
# University -> Department
# Teacher
# FacultyMember(Department, Teacher)

# Parent Class
class University:

    # Constructor
    def __init__(self, university_name, **kwargs):
        super().__init__(**kwargs)
        self.__university_name = university_name

    # Getter
    def get_university_name(self):
        return f"University : {self.__university_name}"

    # Setter
    def set_university_name(self, university_name):
        self.__university_name = university_name
        return "University Name Updated Successfully"

    # Method
    def university_info(self):
        print("University provides higher education.")

# Child Class
class Department(University):

    # Constructor
    def __init__(self, department_name, **kwargs):
        super().__init__(**kwargs)
        self.__department_name = department_name

    # Getter
    def get_department_name(self):
        return f"Department : {self.__department_name}"

    # Setter
    def set_department_name(self, department_name):
        self.__department_name = department_name
        return "Department Name Updated Successfully"

    # Method
    def department_info(self):
        print("Department manages academic programs.")

# Parent Class
class Teacher:

    # Constructor
    def __init__(self, teacher_name, **kwargs):
        super().__init__(**kwargs)
        self.__teacher_name = teacher_name

    # Getter
    def get_teacher_name(self):
        return f"Teacher : {self.__teacher_name}"

    # Setter
    def set_teacher_name(self, teacher_name):
        self.__teacher_name = teacher_name
        return "Teacher Name Updated Successfully"

    # Method
    def teach(self):
        print("Teacher is teaching students.")


# Child Class
class FacultyMember(Department, Teacher):

    # Constructor
    def __init__(self, employee_id, university_name, department_name, teacher_name):
        super().__init__(
            university_name=university_name,
            department_name=department_name,
            teacher_name=teacher_name
        )
        self.__employee_id = employee_id

    # Getter
    def get_employee_id(self):
        return f"Employee ID : {self.__employee_id}"

    # Setter
    def set_employee_id(self, employee_id):
        self.__employee_id = employee_id
        return "Employee ID Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_employee_id())
        print(self.get_university_name())
        print(self.get_department_name())
        print(self.get_teacher_name())

        self.university_info()
        self.department_info()
        self.teach()


# Creating Object
obj = FacultyMember(employee_id="EMP101",university_name="IIT Delhi",department_name="Computer Science",teacher_name="Dr. Sharma")

# Calling Methods
print("********** Faculty Details **********")
obj.show_details()

print()

print(obj.set_university_name("NIT Nagpur"))
print(obj.set_department_name("Artificial Intelligence"))
print(obj.set_teacher_name("Dr. Mehta"))
print(obj.set_employee_id("EMP201"))

print()

print("********** Updated Details **********")
obj.show_details()
```

```python
# 2. Create Person, Employee, Student and TeachingAssiatant using hybride inheritance.
#        Person
#       /      \
# Employee    Student
#       \      /
#   TeachingAssistant


# Parent Class
class Person:

    # Constructor
    def __init__(self, name, age, **kwargs):
        super().__init__(**kwargs)
        self.__name = name
        self.__age = age

    # Getter
    def get_name(self):
        return f"Name : {self.__name}"

    def get_age(self):
        return f"Age : {self.__age}"

    # Setter
    def set_name(self, name):
        self.__name = name
        return "Name Updated Successfully"

    def set_age(self, age):
        self.__age = age
        return "Age Updated Successfully"

    # Method
    def person_info(self):
        print("Person Information Available.")

# Child Class
class Employee(Person):

    # Constructor
    def __init__(self, salary, **kwargs):
        super().__init__(**kwargs)
        self.__salary = salary

    # Getter
    def get_salary(self):
        return f"Salary : ₹{self.__salary}"

    # Setter
    def set_salary(self, salary):
        self.__salary = salary
        return "Salary Updated Successfully"

    # Method
    def employee_work(self):
        print("Employee is Performing Official Duties.")


# Child Class
class Student(Person):

    # Constructor
    def __init__(self, marks, **kwargs):
        super().__init__(**kwargs)
        self.__marks = marks

    # Getter
    def get_marks(self):
        return f"Marks : {self.__marks}"

    # Setter
    def set_marks(self, marks):
        self.__marks = marks
        return "Marks Updated Successfully"

    # Method
    def study(self):
        print("Student is Studying.")

# Child Class
class TeachingAssistant(Employee, Student):

    # Constructor
    def __init__(self, department, name, age, salary, marks):
        super().__init__(
            name=name,
            age=age,
            salary=salary,
            marks=marks
        )
        self.__department = department

    # Getter
    def get_department(self):
        return f"Department : {self.__department}"

    # Setter
    def set_department(self, department):
        self.__department = department
        return "Department Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_name())
        print(self.get_age())
        print(self.get_salary())
        print(self.get_marks())
        print(self.get_department())

        self.person_info()
        self.employee_work()
        self.study()


# Creating Object
obj = TeachingAssistant(department="Computer Science",name="Sumit",age=21,salary=35000,marks=92)

# Calling Methods
print("********** Teaching Assistant Details **********")
obj.show_details()

print()

print(obj.set_name("Amit"))
print(obj.set_age(24))
print(obj.set_salary(42000))
print(obj.set_marks(96))
print(obj.set_department("Artificial Intelligence"))

print()

print("********** Updated Details **********")
obj.show_details()
```

```python
# 3. Create Vehicle → Car, Electric → ElectricCar(ElectricCar inherits from both).
#        Vehicle
#           |
#          Car
#      Electric
#          |
#     ElectricCar(Car, Electric)


# Parent Class
class Vehicle:

    # Constructor
    def __init__(self, brand, **kwargs):
        super().__init__(**kwargs)
        self.__brand = brand

    # Getter
    def get_brand(self):
        return f"Brand : {self.__brand}"

    # Setter
    def set_brand(self, brand):
        self.__brand = brand
        return "Brand Updated Successfully"

    # Method
    def start(self):
        print("Vehicle Started Successfully.")


# Child Class
class Car(Vehicle):

    # Constructor
    def __init__(self, model, **kwargs):
        super().__init__(**kwargs)
        self.__model = model

    # Getter
    def get_model(self):
        return f"Model : {self.__model}"

    # Setter
    def set_model(self, model):
        self.__model = model
        return "Model Updated Successfully"

    # Method
    def drive(self):
        print("Car is Ready to Drive.")


# Parent Class
class Electric:

    # Constructor
    def __init__(self, battery_capacity, **kwargs):
        super().__init__(**kwargs)
        self.__battery_capacity = battery_capacity

    # Getter
    def get_battery_capacity(self):
        return f"Battery Capacity : {self.__battery_capacity} kWh"

    # Setter
    def set_battery_capacity(self, battery_capacity):
        self.__battery_capacity = battery_capacity
        return "Battery Capacity Updated Successfully"

    # Method
    def battery(self):
        print("Battery is Fully Charged.")


# Child Class
class ElectricCar(Car, Electric):

    # Constructor
    def __init__(self, color, brand, model, battery_capacity):
        super().__init__(
            brand=brand,
            model=model,
            battery_capacity=battery_capacity
        )
        self.__color = color

    # Getter
    def get_color(self):
        return f"Color : {self.__color}"

    # Setter
    def set_color(self, color):
        self.__color = color
        return "Color Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_brand())
        print(self.get_model())
        print(self.get_battery_capacity())
        print(self.get_color())

        self.start()
        self.drive()
        self.battery()


# Creating Object
obj = ElectricCar(color="White",brand="Tata",model="Nexon EV",battery_capacity=40)

# Calling Methods
print("********** Electric Car Details **********")
obj.show_details()

print()

print(obj.set_brand("Mahindra"))
print(obj.set_model("BE 6"))
print(obj.set_battery_capacity(60))
print(obj.set_color("Black"))

print()

print("********** Updated Details **********")
obj.show_details()
```

```python
# 4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
#    A
#    |
#    B

#    C

# Parent Class
class Hospital:

    # Constructor
    def __init__(self, hospital_name, **kwargs):
        super().__init__(**kwargs)
        self.__hospital_name = hospital_name

    # Getter
    def get_hospital_name(self):
        return f"Hospital : {self.__hospital_name}"

    # Setter
    def set_hospital_name(self, hospital_name):
        self.__hospital_name = hospital_name
        return "Hospital Name Updated Successfully"

    # Method
    def work(self):
        print("Hospital provides healthcare services.")


# Child Class
class Doctor(Hospital):

    # Constructor
    def __init__(self, doctor_name, **kwargs):
        super().__init__(**kwargs)
        self.__doctor_name = doctor_name

    # Getter
    def get_doctor_name(self):
        return f"Doctor : {self.__doctor_name}"

    # Setter
    def set_doctor_name(self, doctor_name):
        self.__doctor_name = doctor_name
        return "Doctor Name Updated Successfully"

    # Overriding Method
    def work(self):
        print("Doctor is Treating Patients.")


# Parent Class
class Nurse:

    # Constructor
    def __init__(self, nurse_name, **kwargs):
        super().__init__(**kwargs)
        self.__nurse_name = nurse_name

    # Getter
    def get_nurse_name(self):
        return f"Nurse : {self.__nurse_name}"

    # Setter
    def set_nurse_name(self, nurse_name):
        self.__nurse_name = nurse_name
        return "Nurse Name Updated Successfully"

    # Overriding Method
    def work(self):
        print("Nurse is Caring for Patients.")


# Child Class
class Surgeon(Doctor, Nurse):

    # Constructor
    def __init__(self, specialization, hospital_name, doctor_name, nurse_name):
        super().__init__(
            hospital_name=hospital_name,
            doctor_name=doctor_name,
            nurse_name=nurse_name
        )
        self.__specialization = specialization

    # Getter
    def get_specialization(self):
        return f"Specialization : {self.__specialization}"

    # Setter
    def set_specialization(self, specialization):
        self.__specialization = specialization
        return "Specialization Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_hospital_name())
        print(self.get_doctor_name())
        print(self.get_nurse_name())
        print(self.get_specialization())

        print("\nCalling work()")
        self.work()


# Creating Object
obj = Surgeon(specialization="Cardiac Surgeon",hospital_name="AIIMS Delhi",doctor_name="Dr. Sharma",nurse_name="Ms. Priya")

# Calling Methods
print("********** Surgeon Details **********")
obj.show_details()

print("\n********** Method Resolution Order **********")
for cls in Surgeon.mro():
    print(cls.__name__)
```

```python
# 5. Create class A, B, C, D and solve the Diamond Problem using super().
#          Company
#         /       \
#    Manager    Developer
#         \       /
#        TeamLead


# Parent Class
class Company:

    # Constructor
    def __init__(self, company_name, **kwargs):
        super().__init__(**kwargs)
        self.__company_name = company_name
        print("Company Constructor Called")

    # Getter
    def get_company_name(self):
        return f"Company : {self.__company_name}"

    # Setter
    def set_company_name(self, company_name):
        self.__company_name = company_name
        return "Company Name Updated Successfully"

    # Method
    def company_info(self):
        print("Company Provides IT Services.")


# Child Class
class Manager(Company):

    # Constructor
    def __init__(self, manager_name, **kwargs):
        super().__init__(**kwargs)
        self.__manager_name = manager_name
        print("Manager Constructor Called")

    # Getter
    def get_manager_name(self):
        return f"Manager : {self.__manager_name}"

    # Setter
    def set_manager_name(self, manager_name):
        self.__manager_name = manager_name
        return "Manager Name Updated Successfully"

    # Method
    def manage(self):
        print("Manager is Managing the Team.")


# Child Class
class Developer(Company):

    # Constructor
    def __init__(self, developer_name, **kwargs):
        super().__init__(**kwargs)
        self.__developer_name = developer_name
        print("Developer Constructor Called")

    # Getter
    def get_developer_name(self):
        return f"Developer : {self.__developer_name}"

    # Setter
    def set_developer_name(self, developer_name):
        self.__developer_name = developer_name
        return "Developer Name Updated Successfully"

    # Method
    def develop(self):
        print("Developer is Writing Code.")


# Child Class
class TeamLead(Manager, Developer):

    # Constructor
    def __init__(self, team_name, company_name, manager_name, developer_name):
        super().__init__(
            company_name=company_name,
            manager_name=manager_name,
            developer_name=developer_name
        )
        self.__team_name = team_name
        print("TeamLead Constructor Called")

    # Getter
    def get_team_name(self):
        return f"Team : {self.__team_name}"

    # Setter
    def set_team_name(self, team_name):
        self.__team_name = team_name
        return "Team Name Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_company_name())
        print(self.get_manager_name())
        print(self.get_developer_name())
        print(self.get_team_name())

        self.company_info()
        self.manage()
        self.develop()


# Creating Object
obj = TeamLead(team_name="AI Team",company_name="OpenAI",manager_name="Rahul",developer_name="Ananya")

print()

# Calling Methods
print("********** Team Lead Details **********")
obj.show_details()

print()

print(obj.set_company_name("Google"))
print(obj.set_manager_name("Amit"))
print(obj.set_developer_name("Priya"))
print(obj.set_team_name("ML Team"))

print()

print("********** Updated Details **********")
obj.show_details()

print()

print("********** Method Resolution Order **********")
for cls in TeamLead.mro():
    print(cls.__name__)
```

```python
# 6. Create a banking system using hybrid inheritance.
#            Account
#               |
#       SavingsAccount
#       LoanAccount
#            |
#   SavingsLoanAccount


# Parent Class
class Account:

    # Constructor
    def __init__(self, account_number, **kwargs):
        super().__init__(**kwargs)
        self.__account_number = account_number

    # Getter
    def get_account_number(self):
        return f"Account Number : {self.__account_number}"

    # Setter
    def set_account_number(self, account_number):
        self.__account_number = account_number
        return "Account Number Updated Successfully"

    # Method
    def account_info(self):
        print("Bank Account is Active.")


# Child Class
class SavingsAccount(Account):

    # Constructor
    def __init__(self, balance, **kwargs):
        super().__init__(**kwargs)
        self.__balance = balance

    # Getter
    def get_balance(self):
        return f"Balance : ₹{self.__balance}"

    # Setter
    def set_balance(self, balance):
        self.__balance = balance
        return "Balance Updated Successfully"

    # Method
    def deposit(self):
        print("Amount Deposited Successfully.")


# Parent Class
class LoanAccount:

    # Constructor
    def __init__(self, loan_amount, **kwargs):
        super().__init__(**kwargs)
        self.__loan_amount = loan_amount

    # Getter
    def get_loan_amount(self):
        return f"Loan Amount : ₹{self.__loan_amount}"

    # Setter
    def set_loan_amount(self, loan_amount):
        self.__loan_amount = loan_amount
        return "Loan Amount Updated Successfully"

    # Method
    def pay_emi(self):
        print("Loan EMI Paid Successfully.")


# Child Class
class SavingsLoanAccount(SavingsAccount, LoanAccount):

    # Constructor
    def __init__(self, customer_name, account_number, balance, loan_amount):
        super().__init__(
            account_number=account_number,
            balance=balance,
            loan_amount=loan_amount
        )
        self.__customer_name = customer_name

    # Getter
    def get_customer_name(self):
        return f"Customer Name : {self.__customer_name}"

    # Setter
    def set_customer_name(self, customer_name):
        self.__customer_name = customer_name
        return "Customer Name Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_customer_name())
        print(self.get_account_number())
        print(self.get_balance())
        print(self.get_loan_amount())

        self.account_info()
        self.deposit()
        self.pay_emi()


# Creating Object
obj = SavingsLoanAccount(customer_name="Sparsh",account_number="SBI123456789",balance=150000,loan_amount=500000)

# Calling Methods
print("********** Savings Loan Account Details **********")
obj.show_details()

print()

print(obj.set_customer_name("Rahul"))
print(obj.set_account_number("HDFC987654321"))
print(obj.set_balance(200000))
print(obj.set_loan_amount(450000))

print()

print("********** Updated Details **********")
obj.show_details()
```

```python
# 7. Create Shape, Color, Circle, and ColoredCircle.
#            Shape
#               |
#            Circle
# Color
#       |
# ColoredCircle(Circle, Color)


# Parent Class
class Shape:

    # Constructor
    def __init__(self, radius, **kwargs):
        super().__init__(**kwargs)
        self.__radius = radius

    # Getter
    def get_radius(self):
        return f"Radius : {self.__radius}"

    # Setter
    def set_radius(self, radius):
        self.__radius = radius
        return "Radius Updated Successfully"

    # Method
    def area(self):
        print(f"Area : {3.14 * self.__radius * self.__radius}")


# Child Class
class Circle(Shape):

    # Constructor
    def __init__(self, circle_name, **kwargs):
        super().__init__(**kwargs)
        self.__circle_name = circle_name

    # Getter
    def get_circle_name(self):
        return f"Shape Name : {self.__circle_name}"

    # Setter
    def set_circle_name(self, circle_name):
        self.__circle_name = circle_name
        return "Shape Name Updated Successfully"

    # Method
    def draw(self):
        print("Circle Drawn Successfully.")


# Parent Class
class Color:

    # Constructor
    def __init__(self, color, **kwargs):
        super().__init__(**kwargs)
        self.__color = color

    # Getter
    def get_color(self):
        return f"Color : {self.__color}"

    # Setter
    def set_color(self, color):
        self.__color = color
        return "Color Updated Successfully"

    # Method
    def fill_color(self):
        print(f"{self.__color} Color Applied Successfully.")


# Child Class
class ColoredCircle(Circle, Color):

    # Constructor
    def __init__(self, designer, radius, circle_name, color):
        super().__init__(
            radius=radius,
            circle_name=circle_name,
            color=color
        )
        self.__designer = designer

    # Getter
    def get_designer(self):
        return f"Designer : {self.__designer}"

    # Setter
    def set_designer(self, designer):
        self.__designer = designer
        return "Designer Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_designer())
        print(self.get_circle_name())
        print(self.get_radius())
        print(self.get_color())

        self.draw()
        self.area()
        self.fill_color()


# Creating Object
obj = ColoredCircle(designer="Raunak",radius=7,circle_name="Logo Circle",color="Blue")

# Calling Methods
print("********** Colored Circle Details **********")
obj.show_details()

print()

print(obj.set_designer("Rahul"))
print(obj.set_circle_name("Profile Logo"))
print(obj.set_radius(10))
print(obj.set_color("Green"))

print()

print("********** Updated Details **********")
obj.show_details()
```

```python
# 8. Create construction at every level.
#             User
#               |
#           Customer
# PremiumMember
#        |
# PremiumCustomer(Customer, PremiumMember)


# Parent Class
class User:

    # Constructor
    def __init__(self, username, **kwargs):
        super().__init__(**kwargs)
        self.__username = username
        print("User Constructor Called")

    # Getter
    def get_username(self):
        return f"Username : {self.__username}"

    # Setter
    def set_username(self, username):
        self.__username = username
        return "Username Updated Successfully"


# Child Class
class Customer(User):

    # Constructor
    def __init__(self, customer_id, **kwargs):
        super().__init__(**kwargs)
        self.__customer_id = customer_id
        print("Customer Constructor Called")

    # Getter
    def get_customer_id(self):
        return f"Customer ID : {self.__customer_id}"

    # Setter
    def set_customer_id(self, customer_id):
        self.__customer_id = customer_id
        return "Customer ID Updated Successfully"


# Parent Class
class PremiumMember:

    # Constructor
    def __init__(self, membership, **kwargs):
        super().__init__(**kwargs)
        self.__membership = membership
        print("PremiumMember Constructor Called")

    # Getter
    def get_membership(self):
        return f"Membership : {self.__membership}"

    # Setter
    def set_membership(self, membership):
        self.__membership = membership
        return "Membership Updated Successfully"


# Child Class
class PremiumCustomer(Customer, PremiumMember):

    # Constructor
    def __init__(self, reward_points,
                 username,
                 customer_id,
                 membership):

        super().__init__(
            username=username,
            customer_id=customer_id,
            membership=membership
        )

        self.__reward_points = reward_points
        print("PremiumCustomer Constructor Called")

    # Getter
    def get_reward_points(self):
        return f"Reward Points : {self.__reward_points}"

    # Setter
    def set_reward_points(self, reward_points):
        self.__reward_points = reward_points
        return "Reward Points Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_username())
        print(self.get_customer_id())
        print(self.get_membership())
        print(self.get_reward_points())


# Creating Object
obj = PremiumCustomer(reward_points=2500, username="Shubhangi",customer_id="C101",membership="Gold")

print()

# Calling Methods
print("********** Premium Customer Details **********")
obj.show_details()

print()

print(obj.set_username("Rahul"))
print(obj.set_customer_id("C202"))
print(obj.set_membership("Platinum"))
print(obj.set_reward_points(5000))

print()

print("********** Updated Details **********")
obj.show_details()

print()

print("********** Method Resolution Order **********")
for cls in PremiumCustomer.mro():
    print(cls.__name__)
```

```python
# 9. Override method in intermediate classes.
#            School
#               |
#           Teacher
# Principal
#       |
# ClassTeacher(Teacher, Principal)


# Parent Class
class School:

    # Constructor
    def __init__(self, school_name, **kwargs):
        super().__init__(**kwargs)
        self.__school_name = school_name

    # Getter
    def get_school_name(self):
        return f"School : {self.__school_name}"

    # Setter
    def set_school_name(self, school_name):
        self.__school_name = school_name
        return "School Name Updated Successfully"

    # Method
    def show(self):
        print("Welcome to the School.")


# Child Class
class Teacher(School):

    # Constructor
    def __init__(self, teacher_name, **kwargs):
        super().__init__(**kwargs)
        self.__teacher_name = teacher_name

    # Getter
    def get_teacher_name(self):
        return f"Teacher : {self.__teacher_name}"

    # Setter
    def set_teacher_name(self, teacher_name):
        self.__teacher_name = teacher_name
        return "Teacher Name Updated Successfully"

    # Overriding Method
    def show(self):
        print("Teacher is Teaching Students.")


# Parent Class
class Principal:

    # Constructor
    def __init__(self, principal_name, **kwargs):
        super().__init__(**kwargs)
        self.__principal_name = principal_name

    # Getter
    def get_principal_name(self):
        return f"Principal : {self.__principal_name}"

    # Setter
    def set_principal_name(self, principal_name):
        self.__principal_name = principal_name
        return "Principal Name Updated Successfully"

    # Overriding Method
    def show(self):
        print("Principal is Managing the School.")


# Child Class
class ClassTeacher(Teacher, Principal):

    # Constructor
    def __init__(self, class_name,
                 school_name,
                 teacher_name,
                 principal_name):

        super().__init__(
            school_name=school_name,
            teacher_name=teacher_name,
            principal_name=principal_name
        )

        self.__class_name = class_name

    # Getter
    def get_class_name(self):
        return f"Class : {self.__class_name}"

    # Setter
    def set_class_name(self, class_name):
        self.__class_name = class_name
        return "Class Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_school_name())
        print(self.get_teacher_name())
        print(self.get_principal_name())
        print(self.get_class_name())

        print("\nCalling show() Method")
        self.show()


# Creating Object
obj = ClassTeacher(class_name="Class XII",school_name="Prerna College",teacher_name="Mrs. Sharma",principal_name="Mr. Verma")

# Calling Methods
print("********** Class Teacher Details **********")
obj.show_details()

print()

print(obj.set_school_name("Delhi Public School"))
print(obj.set_teacher_name("Mr. Mehta"))
print(obj.set_principal_name("Mrs. Gupta"))
print(obj.set_class_name("Class XI"))

print()

print("********** Updated Details **********")
obj.show_details()

print()

print("********** Method Resolution Order **********")
for cls in ClassTeacher.mro():
    print(cls.__name__)
```

```python
# 10. Create a Device → Phone, Camera → SmartPhone hierarchy.
#            Device
#               |
#            Phone
# Camera
#       |
# SmartPhone(Phone, Camera)


# Parent Class
class Device:

    # Constructor
    def __init__(self, brand, **kwargs):
        super().__init__(**kwargs)
        self.__brand = brand

    # Getter
    def get_brand(self):
        return f"Brand : {self.__brand}"

    # Setter
    def set_brand(self, brand):
        self.__brand = brand
        return "Brand Updated Successfully"

    # Method
    def power_on(self):
        print("Device Powered ON Successfully.")


# Child Class
class Phone(Device):

    # Constructor
    def __init__(self, model, **kwargs):
        super().__init__(**kwargs)
        self.__model = model

    # Getter
    def get_model(self):
        return f"Model : {self.__model}"

    # Setter
    def set_model(self, model):
        self.__model = model
        return "Model Updated Successfully"

    # Method
    def call(self):
        print("Calling a Contact...")


# Parent Class
class Camera:

    # Constructor
    def __init__(self, camera_mp, **kwargs):
        super().__init__(**kwargs)
        self.__camera_mp = camera_mp

    # Getter
    def get_camera_mp(self):
        return f"Camera : {self.__camera_mp} MP"

    # Setter
    def set_camera_mp(self, camera_mp):
        self.__camera_mp = camera_mp
        return "Camera Updated Successfully"

    # Method
    def click_photo(self):
        print("Photo Captured Successfully.")


# Child Class
class SmartPhone(Phone, Camera):

    # Constructor
    def __init__(self, owner_name, brand, model, camera_mp):

        super().__init__(
            brand=brand,
            model=model,
            camera_mp=camera_mp
        )

        self.__owner_name = owner_name

    # Getter
    def get_owner_name(self):
        return f"Owner : {self.__owner_name}"

    # Setter
    def set_owner_name(self, owner_name):
        self.__owner_name = owner_name
        return "Owner Name Updated Successfully"

    # Display Method
    def show_details(self):

        print(self.get_owner_name())
        print(self.get_brand())
        print(self.get_model())
        print(self.get_camera_mp())

        self.power_on()
        self.call()
        self.click_photo()


# Creating Object
obj = SmartPhone(owner_name="Ayushi",brand="Samsung",model="Galaxy S25",camera_mp=200)

# Calling Methods
print("********** SmartPhone Details **********")
obj.show_details()

print()

print(obj.set_owner_name("Rahul"))
print(obj.set_brand("Apple"))
print(obj.set_model("iPhone 17"))
print(obj.set_camera_mp(48))

print()

print("********** Updated Details **********")
obj.show_details()

print()

print("********** Method Resolution Order **********")
for cls in SmartPhone.mro():
    print(cls.__name__)
```

```python
# 19. Create a hybrid inheritance sttructure and apply Duck Typing.
# Hybrid Inheritance + Duck Typing
#              Person
#                 |
#             Employee
# Manager
#      |
# TeamLeader(Employee, Manager)


# Parent Class
class Person:

    # Constructor
    def __init__(self, name, **kwargs):
        super().__init__(**kwargs)
        self.__name = name

    # Getter
    def get_name(self):
        return f"Name : {self.__name}"

    # Setter
    def set_name(self, name):
        self.__name = name
        return "Name Updated Successfully"


# Child Class
class Employee(Person):

    # Constructor
    def __init__(self, emp_id, **kwargs):
        super().__init__(**kwargs)
        self.__emp_id = emp_id

    # Getter
    def get_emp_id(self):
        return f"Employee ID : {self.__emp_id}"

    # Setter
    def set_emp_id(self, emp_id):
        self.__emp_id = emp_id
        return "Employee ID Updated Successfully"


# Parent Class
class Manager:

    # Constructor
    def __init__(self, department, **kwargs):
        super().__init__(**kwargs)
        self.__department = department

    # Getter
    def get_department(self):
        return f"Department : {self.__department}"

    # Setter
    def set_department(self, department):
        self.__department = department
        return "Department Updated Successfully"


# Child Class
class TeamLeader(Employee, Manager):

    # Constructor
    def __init__(self, team_name, name, emp_id, department):

        super().__init__(
            name=name,
            emp_id=emp_id,
            department=department
        )

        self.__team_name = team_name

    # Getter
    def get_team_name(self):
        return f"Team : {self.__team_name}"

    # Setter
    def set_team_name(self, team_name):
        self.__team_name = team_name
        return "Team Name Updated Successfully"

    # Common Method
    def show(self):
        print("********** Team Leader **********")
        print(self.get_name())
        print(self.get_emp_id())
        print(self.get_department())
        print(self.get_team_name())


# Unrelated Class
class Freelancer:

    # Constructor
    def __init__(self, freelancer_name, project):
        self.__freelancer_name = freelancer_name
        self.__project = project

    # Getter
    def get_freelancer_name(self):
        return f"Freelancer : {self.__freelancer_name}"

    def get_project(self):
        return f"Project : {self.__project}"

    # Setter
    def set_freelancer_name(self, freelancer_name):
        self.__freelancer_name = freelancer_name
        return "Freelancer Name Updated Successfully"

    def set_project(self, project):
        self.__project = project
        return "Project Updated Successfully"

    # Common Method
    def show(self):
        print("********** Freelancer **********")
        print(self.get_freelancer_name())
        print(self.get_project())


# Duck Typing Function
def display(obj):
    obj.show()


# Creating Objects
leader = TeamLeader(team_name="AI Team",name="Ayushi",emp_id="EMP101",department="Data Science")

freelancer = Freelancer(freelancer_name="Rahul",project="Website Development")


# Calling Methods
display(leader)

print()

display(freelancer)

print()

print(leader.set_name("Sneha"))
print(leader.set_emp_id("EMP202"))
print(leader.set_department("Machine Learning"))
print(leader.set_team_name("GenAI Team"))

print()

print(freelancer.set_freelancer_name("Aman"))
print(freelancer.set_project("Mobile App"))

print()

print("********** Updated Details **********")
display(leader)

print()

display(freelancer)
```

```python
# 20. Create Employee hierarchy using Hybrid inheritance and use Duck Typing.
# Hybrid Inheritance + Duck Typing
#                Person
#                   |
#               Employee
#             /          \
#        Manager         HR
#             \          /
#              TeamLead
#
# Freelancer (Unrelated Class)


# Parent Class
class Person:

    # Constructor
    def __init__(self, name, **kwargs):
        super().__init__(**kwargs)
        self.__name = name

    # Getter
    def get_name(self):
        return f"Name : {self.__name}"

    # Setter
    def set_name(self, name):
        self.__name = name
        return "Name Updated Successfully"


# Child Class
class Employee(Person):

    # Constructor
    def __init__(self, emp_id, **kwargs):
        super().__init__(**kwargs)
        self.__emp_id = emp_id

    # Getter
    def get_emp_id(self):
        return f"Employee ID : {self.__emp_id}"

    # Setter
    def set_emp_id(self, emp_id):
        self.__emp_id = emp_id
        return "Employee ID Updated Successfully"


# Parent Class
class Manager(Employee):

    # Constructor
    def __init__(self, department, **kwargs):
        super().__init__(**kwargs)
        self.__department = department

    # Getter
    def get_department(self):
        return f"Department : {self.__department}"

    # Setter
    def set_department(self, department):
        self.__department = department
        return "Department Updated Successfully"


# Parent Class
class HR(Employee):

    # Constructor
    def __init__(self, location, **kwargs):
        super().__init__(**kwargs)
        self.__location = location

    # Getter
    def get_location(self):
        return f"Location : {self.__location}"

    # Setter
    def set_location(self, location):
        self.__location = location
        return "Location Updated Successfully"


# Child Class
class TeamLead(Manager, HR):

    # Constructor
    def __init__(self,
                 team_name,
                 name,
                 emp_id,
                 department,
                 location):

        super().__init__(
            name=name,
            emp_id=emp_id,
            department=department,
            location=location
        )

        self.__team_name = team_name

    # Getter
    def get_team_name(self):
        return f"Team : {self.__team_name}"

    # Setter
    def set_team_name(self, team_name):
        self.__team_name = team_name
        return "Team Name Updated Successfully"

    # Common Method
    def work(self):
        print("********** Team Lead **********")
        print(self.get_name())
        print(self.get_emp_id())
        print(self.get_department())
        print(self.get_location())
        print(self.get_team_name())
        print("Managing the Development Team.")


# Unrelated Class
class Freelancer:

    # Constructor
    def __init__(self, name, project):
        self.__name = name
        self.__project = project

    # Getter
    def get_name(self):
        return f"Freelancer : {self.__name}"

    def get_project(self):
        return f"Project : {self.__project}"

    # Setter
    def set_name(self, name):
        self.__name = name
        return "Freelancer Name Updated Successfully"

    def set_project(self, project):
        self.__project = project
        return "Project Updated Successfully"

    # Common Method
    def work(self):
        print("********** Freelancer **********")
        print(self.get_name())
        print(self.get_project())
        print("Working on Client Project.")


# Duck Typing Function
def do_work(obj):
    obj.work()


# Creating Objects
lead = TeamLead(team_name="AI Team",name="Ayushi",emp_id="EMP101",department="Data Science",location="Nagpur")

freelancer = Freelancer(name="Rahul",project="E-Commerce Website")

# Calling Methods
do_work(lead)

print()

do_work(freelancer)

print()

print(lead.set_name("Sneha"))
print(lead.set_emp_id("EMP202"))
print(lead.set_department("Machine Learning"))
print(lead.set_location("Pune"))
print(lead.set_team_name("GenAI Team"))

print()

print(freelancer.set_name("Aman"))
print(freelancer.set_project("Portfolio Website"))

print()

print("********** Updated Details **********")

do_work(lead)

print()

do_work(freelancer)

print()

print("********** Method Resolution Order **********")
for cls in TeamLead.mro():
    print(cls.__name__)
```
