```python
# 2. Define class A with method show() and class B with method display(). Create class C that inherits from both and calls both methods.
# Parent 1
class A:

    # Constructor
    def __init__(self, message1, **kwargs):
        super().__init__(**kwargs)
        self.__message1 = message1

    # Getter
    def get_message1(self):
        return f"Message 1 : {self.__message1}"

    # Setter
    def set_message1(self, message1):
        self.__message1 = message1
        return "Message 1 Updated Successfully"

    # Method
    def show(self):
        print("This is Show() Method of Class A")


# Parent 2
class B:

    # Constructor
    def __init__(self, message2, **kwargs):
        super().__init__(**kwargs)
        self.__message2 = message2

    # Getter
    def get_message2(self):
        return f"Message 2 : {self.__message2}"

    # Setter
    def set_message2(self, message2):
        self.__message2 = message2
        return "Message 2 Updated Successfully"

    # Method
    def display(self):
        print("This is Display() Method of Class B")


# Child
class C(A, B):

    # Constructor
    def __init__(self, message1, message2):
        super().__init__(
            message1=message1,
            message2=message2
        )

    # Display Method
    def show_details(self):
        print(self.get_message1())
        print(self.get_message2())
        self.show()
        self.display()


# Creating Object
obj = C(
    message1="Hello from Class A",
    message2="Hello from Class B"
)

# Calling Methods
print("********** Class C Details **********")
obj.show_details()

print()

print(obj.set_message1("Welcome to Python"))
print(obj.set_message2("Welcome to Multiple Inheritance"))

print()

print("********** Updated Details **********")
print(obj.get_message1())
print(obj.get_message2())
```

```python
# 6. Define two classes with same method name. Create a child class that inherits from both and resolve the method conflict.
# Parent 1
class Father:

    # Constructor
    def __init__(self, father_name, **kwargs):
        super().__init__(**kwargs)
        self.__father_name = father_name

    # Getter
    def get_father_name(self):
        return f"Father Name : {self.__father_name}"

    # Setter
    def set_father_name(self, father_name):
        self.__father_name = father_name
        return "Father Name Updated Successfully"

    # Method
    def show(self):
        print("This is Show() Method of Father Class")


# Parent 2
class Mother:

    # Constructor
    def __init__(self, mother_name, **kwargs):
        super().__init__(**kwargs)
        self.__mother_name = mother_name

    # Getter
    def get_mother_name(self):
        return f"Mother Name : {self.__mother_name}"

    # Setter
    def set_mother_name(self, mother_name):
        self.__mother_name = mother_name
        return "Mother Name Updated Successfully"

    # Method
    def show(self):
        print("This is Show() Method of Mother Class")


# Child
class Child(Father, Mother):

    # Constructor
    def __init__(self, child_name, father_name, mother_name):
        super().__init__(
            father_name=father_name,
            mother_name=mother_name
        )
        self.__child_name = child_name

    # Getter
    def get_child_name(self):
        return f"Child Name : {self.__child_name}"

    # Setter
    def set_child_name(self, child_name):
        self.__child_name = child_name
        return "Child Name Updated Successfully"

    # Display Method
    def show_details(self):
        print(self.get_child_name())
        print(self.get_father_name())
        print(self.get_mother_name())

        print("\nMethod Resolution Order (MRO)")
        self.show()            # Calls Father's show() because Father comes first
        Mother.show(self)      # Explicitly calls Mother's show()


# Creating Object
obj = Child(
    child_name="Priya",
    father_name="Rajesh",
    mother_name="Sunita"
)

# Calling Methods
print("********** Child Details **********")
obj.show_details()

print()

print(obj.set_child_name("Ananya"))
print(obj.set_father_name("Ramesh"))
print(obj.set_mother_name("Kavita"))

print()

print("********** Updated Details **********")
print(obj.get_child_name())
print(obj.get_father_name())
print(obj.get_mother_name())
```

```python
# 14. Define class Converter with static method to_celsius(). Create class WeatherApp that inherits from multiple classes including Converter.
# Parent 1
class Converter:

    # Constructor
    def __init__(self, **kwargs):
        super().__init__(**kwargs)

    # Static Method
    @staticmethod
    def to_celsius(fahrenheit):
        return (fahrenheit - 32) * 5 / 9


# Parent 2
class WeatherInfo:

    # Constructor
    def __init__(self, city, **kwargs):
        super().__init__(**kwargs)
        self.__city = city

    # Getter
    def get_city(self):
        return f"City : {self.__city}"

    # Setter
    def set_city(self, city):
        self.__city = city
        return "City Updated Successfully"

    # Method
    def show_message(self):
        print("Weather Information")


# Child
class WeatherApp(Converter, WeatherInfo):

    # Constructor
    def __init__(self, city, fahrenheit):
        super().__init__(city=city)
        self.__fahrenheit = fahrenheit

    # Getter
    def get_fahrenheit(self):
        return f"Temperature (°F) : {self.__fahrenheit}"

    # Setter
    def set_fahrenheit(self, fahrenheit):
        self.__fahrenheit = fahrenheit
        return "Temperature Updated Successfully"

    # Display Method
    def show_details(self):
        self.show_message()
        print(self.get_city())
        print(self.get_fahrenheit())

        celsius = self.to_celsius(self.__fahrenheit)
        print(f"Temperature (°C) : {celsius:.2f}")


# Creating Object
obj = WeatherApp(
    city="Nagpur",
    fahrenheit=95
)

# Calling Methods
print("********** Weather Details **********")
obj.show_details()

print()

print(obj.set_city("Mumbai"))
print(obj.set_fahrenheit(86))

print()

print("********** Updated Details **********")
obj.show_details()
```

```python
# 26. Create two parent classes:
#• Employee with method get_salary()
#• Person with method get_details() Create class Manager that inherits from both. Add a method that combines both parent methods.

# Parent 1
class Employee:

    # Constructor
    def __init__(self, salary, **kwargs):
        super().__init__(**kwargs)
        self.__salary = salary

    # Getter
    def get_salary(self):
        return f"Salary : ₹{self.__salary}"

    # Setter
    def set_salary(self, salary):
        self.__salary = salary
        return "Salary Updated Successfully"

    # Method
    def salary_details(self):
        print(f"Monthly Salary : ₹{self.__salary}")


# Parent 2
class Person:

    # Constructor
    def __init__(self, name, age, **kwargs):
        super().__init__(**kwargs)
        self.__name = name
        self.__age = age

    # Getter
    def get_name(self):
        return f"Name : {self.__name}"

    def get_age(self):
        return f"Age : {self.__age}"

    # Setter
    def set_name(self, name):
        self.__name = name
        return "Name Updated Successfully"

    def set_age(self, age):
        self.__age = age
        return "Age Updated Successfully"

    # Method
    def person_details(self):
        print(f"Employee Name : {self.__name}")
        print(f"Age : {self.__age}")


# Child
class Manager(Employee, Person):

    # Constructor
    def __init__(self, department, salary, name, age):
        super().__init__(
            salary=salary,
            name=name,
            age=age
        )
        self.__department = department

    # Getter
    def get_department(self):
        return f"Department : {self.__department}"

    # Setter
    def set_department(self, department):
        self.__department = department
        return "Department Updated Successfully"

    # Combined Method
    def manager_details(self):
        self.person_details()
        self.salary_details()
        print(self.get_department())


# Creating Object
obj = Manager(
    department="Data Science",
    salary=80000,
    name="Ayushi",
    age=21
)

# Calling Methods
print("********** Manager Details **********")
obj.manager_details()

print()

print(obj.set_name("Rahul"))
print(obj.set_age(25))
print(obj.set_salary(95000))
print(obj.set_department("Artificial Intelligence"))

print()

print("********** Updated Details **********")
obj.manager_details()
```

```python
# 27. Create abstract class Device with abstract method power_on() and class Network with method connect(). Create class SmartPhone that inherits from both and implements the abstract method.
from abc import ABC, abstractmethod


# Parent 1
class Device(ABC):

    # Constructor
    def __init__(self, brand, **kwargs):
        super().__init__(**kwargs)
        self.__brand = brand

    # Getter
    def get_brand(self):
        return f"Brand : {self.__brand}"

    # Setter
    def set_brand(self, brand):
        self.__brand = brand
        return "Brand Updated Successfully"

    # Abstract Method
    @abstractmethod
    def power_on(self):
        pass


# Parent 2
class Network:

    # Constructor
    def __init__(self, network_type, **kwargs):
        super().__init__(**kwargs)
        self.__network_type = network_type

    # Getter
    def get_network_type(self):
        return f"Network : {self.__network_type}"

    # Setter
    def set_network_type(self, network_type):
        self.__network_type = network_type
        return "Network Updated Successfully"

    # Method
    def connect(self):
        print(f"Connected to {self.__network_type} Network.")


# Child
class SmartPhone(Device, Network):

    # Constructor
    def __init__(self, model, brand, network_type):
        super().__init__(
            brand=brand,
            network_type=network_type
        )
        self.__model = model

    # Getter
    def get_model(self):
        return f"Model : {self.__model}"

    # Setter
    def set_model(self, model):
        self.__model = model
        return "Model Updated Successfully"

    # Implement Abstract Method
    def power_on(self):
        print("SmartPhone Powered On Successfully.")

    # Display Method
    def show_details(self):
        print(self.get_model())
        print(self.get_brand())
        print(self.get_network_type())
        self.power_on()
        self.connect()


# Creating Object
obj = SmartPhone(
    model="Galaxy S25",
    brand="Samsung",
    network_type="5G"
)

# Calling Methods
print("********** SmartPhone Details **********")
obj.show_details()

print()

print(obj.set_model("iPhone 16"))
print(obj.set_brand("Apple"))
print(obj.set_network_type("Wi-Fi"))

print()

print("********** Updated Details **********")
obj.show_details()
```

```python
# 28. Design a system using multiple inheritance:
#• Class Writer
#• Class Editor Create class Author that inherits from both. Add static method to count total authors.

# Parent 1
class Writer:

    # Constructor
    def __init__(self, book_name, **kwargs):
        super().__init__(**kwargs)
        self.__book_name = book_name

    # Getter
    def get_book_name(self):
        return f"Book Name : {self.__book_name}"

    # Setter
    def set_book_name(self, book_name):
        self.__book_name = book_name
        return "Book Name Updated Successfully"

    # Method
    def write(self):
        print("Writing the Book...")


# Parent 2
class Editor:

    # Constructor
    def __init__(self, magazine_name, **kwargs):
        super().__init__(**kwargs)
        self.__magazine_name = magazine_name

    # Getter
    def get_magazine_name(self):
        return f"Magazine : {self.__magazine_name}"

    # Setter
    def set_magazine_name(self, magazine_name):
        self.__magazine_name = magazine_name
        return "Magazine Name Updated Successfully"

    # Method
    def edit(self):
        print("Editing the Content...")


# Child
class Author(Writer, Editor):

    # Static Variable
    total_authors = 0

    # Constructor
    def __init__(self, author_name, book_name, magazine_name):
        super().__init__(
            book_name=book_name,
            magazine_name=magazine_name
        )
        self.__author_name = author_name
        Author.total_authors += 1

    # Getter
    def get_author_name(self):
        return f"Author Name : {self.__author_name}"

    # Setter
    def set_author_name(self, author_name):
        self.__author_name = author_name
        return "Author Name Updated Successfully"

    # Static Method
    @staticmethod
    def count_authors():
        print(f"Total Authors : {Author.total_authors}")

    # Display Method
    def show_details(self):
        print(self.get_author_name())
        print(self.get_book_name())
        print(self.get_magazine_name())
        self.write()
        self.edit()


# Creating Objects
obj1 = Author(
    author_name="Ayushi",
    book_name="Python Programming",
    magazine_name="Tech Today"
)

obj2 = Author(
    author_name="Rahul",
    book_name="Artificial Intelligence",
    magazine_name="AI World"
)

# Calling Methods
print("********** Author 1 Details **********")
obj1.show_details()

print()

print("********** Author 2 Details **********")
obj2.show_details()

print()

Author.count_authors()

print()

print(obj1.set_author_name("Ananya"))
print(obj1.set_book_name("Data Science"))
print(obj1.set_magazine_name("Coding Weekly"))

print()

print("********** Updated Author 1 Details **********")
obj1.show_details()
```

```python
# 29. Create two abstract classes Printable and Scannable. Create class PrinterScanner that inherits from both and implements both abstract methods.
from abc import ABC, abstractmethod


# Parent 1
class Printable(ABC):

    # Constructor
    def __init__(self, document_name, **kwargs):
        super().__init__(**kwargs)
        self.__document_name = document_name

    # Getter
    def get_document_name(self):
        return f"Document : {self.__document_name}"

    # Setter
    def set_document_name(self, document_name):
        self.__document_name = document_name
        return "Document Name Updated Successfully"

    # Abstract Method
    @abstractmethod
    def print_document(self):
        pass


# Parent 2
class Scannable(ABC):

    # Constructor
    def __init__(self, scan_quality, **kwargs):
        super().__init__(**kwargs)
        self.__scan_quality = scan_quality

    # Getter
    def get_scan_quality(self):
        return f"Scan Quality : {self.__scan_quality}"

    # Setter
    def set_scan_quality(self, scan_quality):
        self.__scan_quality = scan_quality
        return "Scan Quality Updated Successfully"

    # Abstract Method
    @abstractmethod
    def scan_document(self):
        pass


# Child
class PrinterScanner(Printable, Scannable):

    # Constructor
    def __init__(self, company_name, document_name, scan_quality):
        super().__init__(
            document_name=document_name,
            scan_quality=scan_quality
        )
        self.__company_name = company_name

    # Getter
    def get_company_name(self):
        return f"Company : {self.__company_name}"

    # Setter
    def set_company_name(self, company_name):
        self.__company_name = company_name
        return "Company Name Updated Successfully"

    # Implement Abstract Method
    def print_document(self):
        print("Document Printed Successfully.")

    # Implement Abstract Method
    def scan_document(self):
        print("Document Scanned Successfully.")

    # Display Method
    def show_details(self):
        print(self.get_company_name())
        print(self.get_document_name())
        print(self.get_scan_quality())
        self.print_document()
        self.scan_document()


# Creating Object
obj = PrinterScanner(
    company_name="HP",
    document_name="Employee_Report.pdf",
    scan_quality="600 DPI"
)

# Calling Methods
print("********** Printer Scanner Details **********")
obj.show_details()

print()

print(obj.set_company_name("Canon"))
print(obj.set_document_name("Student_Report.pdf"))
print(obj.set_scan_quality("1200 DPI"))

print()

print("********** Updated Details **********")
obj.show_details()
```

```python
# 30. Create class A with static method, class B with abstract method, and class C that inherits from both. Implement the abstract method and call the static method.
from abc import ABC, abstractmethod


# Parent 1
class A:

    # Constructor
    def __init__(self, **kwargs):
        super().__init__(**kwargs)

    # Static Method
    @staticmethod
    def company_name():
        print("Company : OpenAI")


# Parent 2
class B(ABC):

    # Constructor
    def __init__(self, employee_name, **kwargs):
        super().__init__(**kwargs)
        self.__employee_name = employee_name

    # Getter
    def get_employee_name(self):
        return f"Employee Name : {self.__employee_name}"

    # Setter
    def set_employee_name(self, employee_name):
        self.__employee_name = employee_name
        return "Employee Name Updated Successfully"

    # Abstract Method
    @abstractmethod
    def work(self):
        pass


# Child
class C(A, B):

    # Constructor
    def __init__(self, employee_name, department):
        super().__init__(employee_name=employee_name)
        self.__department = department

    # Getter
    def get_department(self):
        return f"Department : {self.__department}"

    # Setter
    def set_department(self, department):
        self.__department = department
        return "Department Updated Successfully"

    # Implement Abstract Method
    def work(self):
        print(f"{self.get_employee_name()} is Working in {self.__department} Department.")

    # Display Method
    def show_details(self):
        print(self.get_employee_name())
        print(self.get_department())

        # Calling Abstract Method
        self.work()

        # Calling Static Method
        self.company_name()


# Creating Object
obj = C(
    employee_name="Ayushi",
    department="Data Science"
)

# Calling Methods
print("********** Employee Details **********")
obj.show_details()

print()

print(obj.set_employee_name("Rahul"))
print(obj.set_department("Artificial Intelligence"))

print()

print("********** Updated Details **********")
obj.show_details()
```
